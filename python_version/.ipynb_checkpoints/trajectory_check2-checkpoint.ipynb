
# Spatial PRC1 → rough population CTMC trajectory check

This notebook does **not** run SMC-ABC.

It does this trajectory check:

1. Run the spatial PRC1 Gillespie model many times.
2. Record diagnostic quantities on the observation grid:
   \(N(t)\), \(n_1(t)\), \(n_2(t)\), available sites, rough \(m_i(t)\), and rough \(k_{21,j}(t)\).
3. Estimate rough CTMC parameters
   \[
   R_{01},\quad k_{10},\quad k_{12},\quad k_{21}.
   \]
4. Simulate the mapped population CTMC.
5. Plot the spatial mean trajectory against the mapped CTMC mean trajectory.

The mapping here is intentionally rough and energy-free, matching the idea in the report.

> **Note (bug fix):** `rough_k10_values_for_singly_bound` previously
> doubled the rate when `per_head_rates=True`. A singly bound PRC1 has
> only one attached head, so the molecule-level rate is
> `singly_bound_detachment_rate`, not `2 * singly_bound_detachment_rate`.
> The factor of 2 was correct for k21 (doubly bound has two heads) and
> remains in `rough_k21_values_for_doubly_bound`.


## Rough diagnostic definitions

The report's rough map is:

\[
R_{01}\approx 2 k_{01}^{\mathrm{site}} A^{\mathrm{spatial}},
\qquad
k_{10}\approx k_{10}^{\mathrm{spatial}},
\]

\[
k_{12}\approx k_{12}^{\mathrm{site}}\widehat m,
\qquad
k_{21}\approx \widehat{k}_{21}^{\mathrm{spatial}}.
\]

Here \(\widehat m\) is the average number of available second-head target sites per singly bound PRC1. With microtubule separation \(30\) nm, rest length/radius \(32\) nm, and site spacing \(8\) nm, the rough reachable offsets are \(-1,0,1\), so at most 3 sites.


```python

def make_spatial_state(
    initial_binding_rate_per_site,
    singly_bound_detachment_rate,
    base_double_attachment_rate,
    base_double_detachment_rate,
    cooperativity_energy=0.0,
    enable_initial_attach_cooperativity=False,
    enable_second_head_attach_cooperativity=False,
    enable_second_head_detach_cooperativity=False,
):
    """
    Same fixed geometry as run_gillespie_prc1 in the uploaded code.
    """
    microtubule_length = 5000
    site_spacing = 8
    microtubule_offset = 0
    spring_constant = 2
    rest_length = 32
    k_B_T = 4.1
    microtubule_separation = 30
    base_hopping_rate = 1

    return State(
        microtubule_length,
        site_spacing,
        microtubule_offset,
        spring_constant,
        rest_length,
        k_B_T,
        microtubule_separation,
        initial_binding_rate_per_site,
        singly_bound_detachment_rate,
        base_double_attachment_rate,
        base_double_detachment_rate,
        base_hopping_rate,
        cooperativity_energy,
        enable_initial_attach_cooperativity,
        enable_second_head_attach_cooperativity,
        enable_second_head_detach_cooperativity,
    )


def count_second_head_targets_rough(prc1):
    """
    Energy-free rough count for m_i(t).

    Candidate target sites are closest-1, closest, closest+1.
    We count candidates that are:
      - inside the microtubule,
      - untaken,
      - allowed by the neighbor-ordering constraint from the spatial code.
    """
    state = prc1.state

    if not prc1.is_singly_attached:
        return 0

    closest = int(prc1.closest_index_on_other_side)
    candidate_sites = [closest - 1, closest, closest + 1]

    # Same ordering bounds used in the spatial attachment routine.
    left_allowed = prc1.left_neighbor_opposite_index + 1
    right_allowed = prc1.right_neighbor_opposite_index  # exclusive

    if prc1.top_head_is_attached:
        opposite_untaken = state.bottom_untaken_sites
    elif prc1.bottom_head_is_attached:
        opposite_untaken = state.top_untaken_sites
    else:
        return 0

    count = 0
    for site in candidate_sites:
        if (
            0 <= site < state.num_sites
            and left_allowed <= site < right_allowed
            and site in opposite_untaken
        ):
            count += 1

    return count


def rough_k10_values_for_singly_bound(state, singly, per_head_rates=True):
    """
    Rough no-energy k10_j values.

    A singly bound PRC1 has exactly ONE attached head, so the molecule-level
    single-bound -> free hazard is just singly_bound_detachment_rate, regardless
    of whether per_head_rates is True or False. The per_head_rates flag is kept
    for API compatibility with rough_k21_values_for_doubly_bound (where the
    factor of 2 is correct because doubly bound molecules have two attachable
    heads), but it is intentionally ignored here.
    """
    return np.full(len(singly), state.singly_bound_detachment_rate, dtype=float)


def rough_k21_values_for_doubly_bound(state, doubly, per_head_rates=True):
    """
    Rough no-energy k21_j values.

    A doubly bound PRC1 can lose either head. If base_double_detachment_rate is
    interpreted as a per-head detachment rate, then the total molecule-level
    double-bound -> single-bound hazard is 2 * base_double_detachment_rate.

    If you want to interpret the input as already molecule-level, set per_head_rates=False.
    """
    factor = 2.0 if per_head_rates else 1.0
    return np.full(len(doubly), factor * state.base_double_detachment_rate, dtype=float)


def spatial_diagnostics_rough(state, per_head_detachment_rates=True):
    """
    Record the quantities needed for the rough spatial -> CTMC map.
    """
    singly = list(state.top_attached_prc1) + list(state.bottom_attached_prc1)
    doubly = list(state.doubly_attached_prc1)

    n1 = len(singly)
    n2 = len(doubly)
    N = n1 + n2

    # Average available sites per microtubule.
    A_spatial = 0.5 * (len(state.top_untaken_sites) + len(state.bottom_untaken_sites))

    # m_i(t): number of rough reachable available second-head target sites.
    m_values = np.array([count_second_head_targets_rough(p) for p in singly], dtype=float)

    # Rough no-energy detachment hazards.
    k10_values = rough_k10_values_for_singly_bound(state, singly, per_head_rates=per_head_detachment_rates)
    k21_values = rough_k21_values_for_doubly_bound(state, doubly, per_head_rates=per_head_detachment_rates)

    return {
        "N": float(N),
        "n1": float(n1),
        "n2": float(n2),
        "A_spatial": float(A_spatial),
        "m_values": m_values,
        "m_sum": float(np.sum(m_values)),
        "k10_values": k10_values,
        "k10_sum": float(np.sum(k10_values)),
        "k21_values": k21_values,
        "k21_sum": float(np.sum(k21_values)),
    }
```


## Spatial model runner with diagnostic recording

This is almost the same Gillespie structure as the uploaded `run_gillespie_prc1`, but the statistic function records more than just `len(state)`.


```python

def run_spatial_diagnostics_on_grid(
    initial_binding_rate_per_site,
    singly_bound_detachment_rate,
    base_double_attachment_rate,
    base_double_detachment_rate,
    times_obs,
    cooperativity_energy=0.0,
    enable_initial_attach_cooperativity=False,
    enable_second_head_attach_cooperativity=False,
    enable_second_head_detach_cooperativity=False,
    enable_hopping=True,
    max_steps=200_000,
    per_head_detachment_rates=True,
):
    """
    Run the spatial PRC1 Gillespie model and record diagnostics on times_obs.
    """
    times_obs = np.asarray(times_obs, dtype=float)
    if times_obs.ndim != 1:
        raise ValueError("times_obs must be 1D")
    if len(times_obs) == 0:
        raise ValueError("times_obs cannot be empty")
    if np.any(np.diff(times_obs) < 0):
        raise ValueError("times_obs must be sorted")

    initial_state = make_spatial_state(
        initial_binding_rate_per_site=initial_binding_rate_per_site,
        singly_bound_detachment_rate=singly_bound_detachment_rate,
        base_double_attachment_rate=base_double_attachment_rate,
        base_double_detachment_rate=base_double_detachment_rate,
        cooperativity_energy=cooperativity_energy,
        enable_initial_attach_cooperativity=enable_initial_attach_cooperativity,
        enable_second_head_attach_cooperativity=enable_second_head_attach_cooperativity,
        enable_second_head_detach_cooperativity=enable_second_head_detach_cooperativity,
    )

    def rate_function(state):
        initial_binding_rate = state.initial_binding_rate

        if len(state) == 0:
            return [initial_binding_rate, 0, 0, 0, 0]

        rates = [initial_binding_rate] + [0] * (len(state) - 1)

        for prc1 in state:
            rates.append(prc1.double_attachment_rate)

        for prc1 in state:
            rates.append(prc1.detachment_rate)

        if enable_hopping:
            for prc1 in state:
                rates.append(prc1.total_bottom_hopping_rate)
            for prc1 in state:
                rates.append(prc1.total_top_hopping_rate)

        return rates

    def single_attach_func(state, index):
        state.single_attach_prc1()

    def double_attach_func(state, index):
        state.double_attach_prc1(index)

    def detach_func(state, index):
        state.detach_prc1(index)

    def hop_bottom_func(state, index):
        state.hop_bottom(index)

    def hop_top_func(state, index):
        state.hop_top(index)

    if enable_hopping:
        reaction_functions = [
            single_attach_func,
            double_attach_func,
            detach_func,
            hop_bottom_func,
            hop_top_func,
        ]
    else:
        reaction_functions = [
            single_attach_func,
            double_attach_func,
            detach_func,
        ]

    def statistic_function(state, stat_list, time):
        stat_list.append(spatial_diagnostics_rough(state, per_head_detachment_rates=per_head_detachment_rates))

    def timestep_function(state, time, dt):
        return time + dt

    statistics, times = run_gillespie(
        initial_state,
        end_time=float(times_obs[-1]),
        rate_function=rate_function,
        reaction_functions=reaction_functions,
        statistic_function=statistic_function,
        timestep_function=timestep_function,
        max_timestep=np.inf,
        max_steps=max_steps,
    )

    # Right-continuous sampling onto times_obs.
    idx = np.searchsorted(times, times_obs, side="right") - 1
    idx = np.clip(idx, 0, len(statistics) - 1)

    scalar_keys = ["N", "n1", "n2", "A_spatial", "m_sum", "k10_sum", "k21_sum"]
    out = {"times": times_obs.copy()}
    for key in scalar_keys:
        out[key] = np.array([statistics[i][key] for i in idx], dtype=float)

    # Keep the actual variable-length recorded lists too, useful for inspection.
    out["m_values_by_time"] = [statistics[i]["m_values"] for i in idx]
    out["k10_values_by_time"] = [statistics[i]["k10_values"] for i in idx]
    out["k21_values_by_time"] = [statistics[i]["k21_values"] for i in idx]

    return out
```


## Estimate mapped CTMC parameters from spatial diagnostics

The estimates use molecule-time weighted averages:

\[
\widehat m = \frac{\sum_{r,t}\sum_{i=1}^{n_1(t)}m_i(t)}{\sum_{r,t} n_1(t)},
\qquad
\widehat k_{21} = \frac{\sum_{r,t}\sum_{j=1}^{n_2(t)}k_{21,j}(t)}{\sum_{r,t} n_2(t)}.
\]

Then

\[
\widehat R_{01} = 2k_{01}^{\mathrm{site}}\widehat A,
\qquad
\widehat k_{12}=k_{12}^{\mathrm{site}}\widehat m.
\]


```python

def estimate_ctmc_from_spatial_runs(
    initial_binding_rate_per_site,
    singly_bound_detachment_rate,
    base_double_attachment_rate,
    base_double_detachment_rate,
    times_obs,
    n_reps=20,
    seed=1,
    verbose=True,
    **spatial_kwargs,
):
    """
    Run spatial simulations and estimate rough CTMC parameters.
    """
    rng = np.random.default_rng(seed)
    spatial_runs = []

    for r in range(n_reps):
        # Uploaded spatial code uses np.random directly.
        np.random.seed(int(rng.integers(0, 2**31 - 1)))

        if verbose:
            print(f"Spatial run {r + 1}/{n_reps}")

        run = run_spatial_diagnostics_on_grid(
            initial_binding_rate_per_site=initial_binding_rate_per_site,
            singly_bound_detachment_rate=singly_bound_detachment_rate,
            base_double_attachment_rate=base_double_attachment_rate,
            base_double_detachment_rate=base_double_detachment_rate,
            times_obs=times_obs,
            **spatial_kwargs,
        )
        spatial_runs.append(run)

    A_all = np.concatenate([run["A_spatial"] for run in spatial_runs])
    n1_all = np.concatenate([run["n1"] for run in spatial_runs])
    n2_all = np.concatenate([run["n2"] for run in spatial_runs])
    m_sum_all = np.concatenate([run["m_sum"] for run in spatial_runs])
    k10_sum_all = np.concatenate([run["k10_sum"] for run in spatial_runs])
    k21_sum_all = np.concatenate([run["k21_sum"] for run in spatial_runs])

    total_n1 = np.sum(n1_all)
    total_n2 = np.sum(n2_all)

    A_hat = float(np.mean(A_all))
    m_hat = float(np.sum(m_sum_all) / total_n1) if total_n1 > 0 else 0.0
    k10_hat = float(np.sum(k10_sum_all) / total_n1) if total_n1 > 0 else 0.0
    k21_hat = float(np.sum(k21_sum_all) / total_n2) if total_n2 > 0 else 0.0

    R01_hat = float(2.0 * initial_binding_rate_per_site * A_hat)
    k12_hat = float(base_double_attachment_rate * m_hat)

    theta_hat = {
        "R01": R01_hat,
        "k10": k10_hat,
        "k12": k12_hat,
        "k21": k21_hat,
        "A_hat": A_hat,
        "m_hat": m_hat,
        "total_n1_obs": float(total_n1),
        "total_n2_obs": float(total_n2),
    }

    return spatial_runs, theta_hat
```


## Population CTMC simulator

This simulates the mapped CTMC:

\[
\emptyset \to n_1, \quad n_1\to \emptyset, \quad n_1\to n_2, \quad n_2\to n_1.
\]


```python

def simulate_population_ctmc_on_grid(theta, times_obs, seed=None):
    """
    Population CTMC:
        empty -> n1      rate R01
        n1 -> empty      rate k10 n1
        n1 -> n2         rate k12 n1
        n2 -> n1         rate k21 n2
    """
    R01 = float(theta["R01"])
    k10 = float(theta["k10"])
    k12 = float(theta["k12"])
    k21 = float(theta["k21"])

    rng = np.random.default_rng(seed)
    times_obs = np.asarray(times_obs, dtype=float)

    n1 = 0
    n2 = 0
    t = 0.0

    n1_out = np.zeros(len(times_obs), dtype=float)
    n2_out = np.zeros(len(times_obs), dtype=float)
    N_out = np.zeros(len(times_obs), dtype=float)

    grid_idx = 0

    while grid_idx < len(times_obs) and times_obs[grid_idx] <= 0:
        n1_out[grid_idx] = n1
        n2_out[grid_idx] = n2
        N_out[grid_idx] = n1 + n2
        grid_idx += 1

    while grid_idx < len(times_obs):
        rates = np.array([
            R01,
            k10 * n1,
            k12 * n1,
            k21 * n2,
        ], dtype=float)

        a0 = rates.sum()
        if a0 <= 0:
            n1_out[grid_idx:] = n1
            n2_out[grid_idx:] = n2
            N_out[grid_idx:] = n1 + n2
            break

        dt = rng.exponential(1.0 / a0)
        t_next = t + dt

        while grid_idx < len(times_obs) and times_obs[grid_idx] < t_next:
            n1_out[grid_idx] = n1
            n2_out[grid_idx] = n2
            N_out[grid_idx] = n1 + n2
            grid_idx += 1

        if grid_idx >= len(times_obs):
            break

        t = t_next
        u = rng.random() * a0

        if u < rates[0]:
            n1 += 1
        elif u < rates[0] + rates[1]:
            if n1 > 0:
                n1 -= 1
        elif u < rates[0] + rates[1] + rates[2]:
            if n1 > 0:
                n1 -= 1
                n2 += 1
        else:
            if n2 > 0:
                n2 -= 1
                n1 += 1

    return {
        "times": times_obs,
        "n1": n1_out,
        "n2": n2_out,
        "N": N_out,
    }


def simulate_population_ctmc_many(theta, times_obs, n_reps=20, seed=123):
    rng = np.random.default_rng(seed)
    runs = []
    for _ in range(n_reps):
        run_seed = int(rng.integers(0, 2**31 - 1))
        runs.append(simulate_population_ctmc_on_grid(theta, times_obs, seed=run_seed))
    return runs
```


## Plotting and summary helpers


```python

def stack_runs(runs, key):
    return np.vstack([run[key] for run in runs])


def summarize_trajectories(runs, key):
    arr = stack_runs(runs, key)
    return {
        "mean": arr.mean(axis=0),
        "median": np.median(arr, axis=0),
        "lo": np.percentile(arr, 5, axis=0),
        "hi": np.percentile(arr, 95, axis=0),
    }


def plot_spatial_vs_ctmc(spatial_runs, ctmc_runs, times_obs, key="N", title=None, save_path=None):
    spatial = summarize_trajectories(spatial_runs, key)
    ctmc = summarize_trajectories(ctmc_runs, key)

    if title is None:
        title = f"Spatial model vs mapped CTMC: {key}"

    fig, ax = plt.subplots(figsize=(7, 4))

    ax.plot(times_obs, spatial["mean"], label=f"Spatial mean {key}")
    ax.fill_between(times_obs, spatial["lo"], spatial["hi"], alpha=0.2, label="Spatial 5-95%")

    ax.plot(times_obs, ctmc["mean"], "--", label=f"Mapped CTMC mean {key}")
    ax.fill_between(times_obs, ctmc["lo"], ctmc["hi"], alpha=0.2, label="CTMC 5-95%")

    ax.set_xlabel("time")
    ax.set_ylabel(key)
    ax.set_title(title)
    ax.legend()
    fig.tight_layout()

    if save_path is not None:
        fig.savefig(save_path, dpi=200, bbox_inches="tight")
        print("Saved figure to", save_path)

    plt.show()
    return fig, ax


def plot_n1_n2_N(spatial_runs, ctmc_runs, times_obs, output_dir=None):
    if output_dir is not None:
        output_dir = Path(output_dir)
        output_dir.mkdir(exist_ok=True, parents=True)

    for key in ["N", "n1", "n2"]:
        save_path = None if output_dir is None else output_dir / f"spatial_vs_ctmc_{key}.png"
        plot_spatial_vs_ctmc(spatial_runs, ctmc_runs, times_obs, key=key, save_path=save_path)


def print_theta(theta_hat):
    print("Mapped CTMC parameters:")
    for key in ["R01", "k10", "k12", "k21", "A_hat", "m_hat", "total_n1_obs", "total_n2_obs"]:
        print(f"  {key:>12s} = {theta_hat[key]:.8g}")
```

```python
# ------------------------------------------------------------
# New plotting block: two-color trajectories + two-color means
# ------------------------------------------------------------
import numpy as np
import matplotlib.pyplot as plt

def plot_trajectory_and_mean(spatial_runs, ctmc_runs, times_obs, key="N", max_lines=None):
    """
    Make two plots:
      1) many trajectories from spatial and CTMC runs
      2) mean-only comparison

    Uses two colors only:
      spatial = blue
      CTMC    = orange
    """
    spatial_mat = np.vstack([run[key] for run in spatial_runs])
    ctmc_mat = np.vstack([run[key] for run in ctmc_runs])

    if max_lines is not None:
        spatial_mat = spatial_mat[:max_lines]
        ctmc_mat = ctmc_mat[:max_lines]

    if key == "N":
        ylabel = r"$N(t)=n_1(t)+n_2(t)$"
        title_name = r"$N(t)$"
    elif key == "n1":
        ylabel = r"$n_1(t)$"
        title_name = r"$n_1(t)$"
    elif key == "n2":
        ylabel = r"$n_2(t)$"
        title_name = r"$n_2(t)$"
    else:
        ylabel = key
        title_name = key

    spatial_color = "tab:blue"
    ctmc_color = "tab:orange"

    # --------------------------------------------------------
    # Plot 1: many trajectories, two colors
    # --------------------------------------------------------
    plt.figure(figsize=(8, 5))

    for i in range(spatial_mat.shape[0]):
        label = "Spatial trajectories" if i == 0 else None
        plt.plot(
            times_obs,
            spatial_mat[i],
            color=spatial_color,
            alpha=0.45,
            linewidth=1,
            label=label,
        )

    for i in range(ctmc_mat.shape[0]):
        label = "Population trajectories" if i == 0 else None
        plt.plot(
            times_obs,
            ctmc_mat[i],
            color=ctmc_color,
            alpha=0.45,
            linewidth=1,
            label=label,
        )

    plt.xlabel("time")
    plt.ylabel(ylabel)
    plt.title(f"Trajectories {title_name}")
    plt.legend()
    plt.tight_layout()
    plt.show()

    # --------------------------------------------------------
    # Plot 2: mean only, two colors
    # --------------------------------------------------------
    spatial_mean = spatial_mat.mean(axis=0)
    ctmc_mean = ctmc_mat.mean(axis=0)

    plt.figure(figsize=(8, 5))

    plt.plot(
        times_obs,
        spatial_mean,
        color=spatial_color,
        linewidth=2.5,
        label="Spatial mean",
    )

    plt.plot(
        times_obs,
        ctmc_mean,
        color=ctmc_color,
        linewidth=2.5,
        label="Population mean",
    )

    plt.xlabel("time")
    plt.ylabel(ylabel)
    plt.title(f"Mean trajectory {title_name}")
    plt.legend()
    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# Run plot for total occupancy
# ------------------------------------------------------------
plot_trajectory_and_mean(spatial_runs, ctmc_runs, times_obs, key="N", max_lines=None)

# Optional:
plot_trajectory_and_mean(spatial_runs, ctmc_runs, times_obs, key="n1", max_lines=None)
plot_trajectory_and_mean(spatial_runs, ctmc_runs, times_obs, key="n2", max_lines=None)
```
