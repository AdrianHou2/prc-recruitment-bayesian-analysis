# One-parameter SMC-ABC for k12 with hopping turned off

This notebook performs a **one-parameter inference** for

- `k12` = `base_double_attachment_rate`

using the spatial PRC1 Gillespie model, with:

- all other kinetic parameters fixed,
- a fixed cooperativity energy,
- `enable_cooperativity=True`,
- `enable_hopping=False`.

Before running SMC-ABC, it performs **one timing check** for a single simulation.

## 1. Load local modules

```python
from pathlib import Path
import importlib.util
import inspect
import sys

def load_module(module_name, filename_candidates):
    search_roots = [
        Path.cwd(),
        Path.cwd() / "python_version",
        Path.cwd().parent,
    ]
    for root in search_roots:
        for name in filename_candidates:
            path = root / name
            if path.exists():
                spec = importlib.util.spec_from_file_location(module_name, path)
                module = importlib.util.module_from_spec(spec)
                sys.modules[module_name] = module
                spec.loader.exec_module(module)
                print(f"Loaded {module_name} from: {path}")
                return module
    raise FileNotFoundError(f"Could not find any of: {filename_candidates}")

gillespie = load_module("gillespie", ["gillespie.py", "gillespie(20).py", "gillespie(7).py"])
prc1 = load_module("prc1", ["prc1.py", "prc1(19).py", "prc1(7).py"])
prc1_state = load_module("prc1_state", ["prc1_state.py", "prc1_state(18).py", "prc1_state(7).py"])
run_gillespie = load_module("run_gillespie", ["run_gillespie.py", "run_gillespie(18).py", "run_gillespie(6).py"])
smc_abc = load_module("smc_abc", ["smc_abc.py", "smc_abc(6).py", "smc-abc.py", "smc-abc(6).py"])

sig = inspect.signature(run_gillespie.run_gillespie_prc1_on_grid)
required_flags = {
    "enable_initial_attach_cooperativity",
    "enable_second_head_attach_cooperativity",
    "enable_second_head_detach_cooperativity",
}
missing = sorted(required_flags - set(sig.parameters))
if missing:
    raise RuntimeError(
        "Your loaded run_gillespie module does not have the new cooperativity flags. "
        f"Missing: {missing}. Make sure you are loading the patched files."
    )

print("Patched cooperativity flags detected successfully.")
```

## 2. Fixed parameters and observation grid

```python
import numpy as np

# fixed model parameters
INIT_BIND = 0.002
K_DET_1 = 0.1
K_DET_2 = 0.1
MAX_STEPS = 200_000

# observation grid
times_obs = np.arange(0.0, 68.0 + 4.0, 4.0)

# thermal scale and fixed cooperativity
k_B_T = 4.1
EPS_FIXED = 2.3 * k_B_T

# true k12 used for synthetic data
K12_TRUE = 5.88

print("times_obs =", times_obs)
print("EPS_FIXED =", EPS_FIXED)
print("K_DET_1 =", K_DET_1)
print("K_DET_2 =", K_DET_2)
print("K12_TRUE =", K12_TRUE)
```

## 3. Define a one-parameter simulator

Here `theta[0]` is `k12 = base_double_attachment_rate`.
Everything else is fixed, and hopping is disabled.

```python
def simulate_one_k12_attach_only(theta, times_obs, seed):
    """
    theta[0] = k12 = base_double_attachment_rate
    All other parameters are fixed.
    Hopping is disabled.
    Cooperativity acts only in second-head attachment.
    """
    import numpy as np
    np.random.seed(int(seed))

    return run_gillespie.run_gillespie_prc1_on_grid(
        initial_binding_rate_per_site=INIT_BIND,
        singly_bound_detachment_rate=K_DET_1,
        base_double_attachment_rate=float(theta[0]),
        base_double_detachment_rate=K_DET_2,
        times_obs=times_obs,
        cooperativity_energy=EPS_FIXED,
        enable_initial_attach_cooperativity=False,
        enable_second_head_attach_cooperativity=True,
        enable_second_head_detach_cooperativity=False,
        enable_hopping=False,
        max_steps=MAX_STEPS,
    )
```

## 4. Time one simulation before doing inference

```python
import time

theta_test = np.array([K12_TRUE], dtype=float)

t0 = time.perf_counter()
test_path = simulate_one_k12_attach_only(theta_test, times_obs, seed=123)
t1 = time.perf_counter()

one_sim_seconds = t1 - t0

print("One simulation path:")
print(test_path)
print("Shape:", test_path.shape)
print(f"Time for one simulation: {one_sim_seconds:.6f} seconds")
```

## 5. Patch the SMC-ABC module to use the one-parameter simulator

```python
smc_abc.simulate_one = simulate_one_k12_attach_only
print("Patched smc_abc.simulate_one successfully.")
print("joblib available:", getattr(smc_abc, "_JOBLIB_AVAILABLE", None))
```

## 6. Generate synthetic observed data

```python
n_obs_reps = 10

y_obs = np.array([
    run_gillespie.run_gillespie_prc1_on_grid(
        initial_binding_rate_per_site=INIT_BIND,
        singly_bound_detachment_rate=K_DET_1,
        base_double_attachment_rate=K12_TRUE,
        base_double_detachment_rate=K_DET_2,
        times_obs=times_obs,
        cooperativity_energy=EPS_FIXED,
        enable_initial_attach_cooperativity=False,
        enable_second_head_attach_cooperativity=True,
        enable_second_head_detach_cooperativity=False,
        enable_hopping=False,
        max_steps=MAX_STEPS,
    )
    for _ in range(n_obs_reps)
], dtype=float)

print("y_obs shape:", y_obs.shape)
print(y_obs)
```

## 7. Use priors centered at the true $k_{12}$ value

This section compares a lognormal prior and a truncated normal prior whose means are both centered at the synthetic true value. The SMC run uses the truncated normal prior directly on the original $k_{12}$ scale.


```python
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt

# --- helper functions for a 1D truncated normal prior ---
def truncnorm_sample_scalar(rng, mu, sd, lower, upper, n):
    out = np.empty(int(n), dtype=float)
    filled = 0
    while filled < int(n):
        draw = rng.normal(loc=float(mu), scale=float(sd), size=max(100, 3 * (int(n) - filled)))
        draw = draw[(draw >= float(lower)) & (draw <= float(upper))]
        m = min(draw.size, int(n) - filled)
        if m > 0:
            out[filled:filled+m] = draw[:m]
            filled += m
    return out


def truncnorm_logpdf_scalar(x, mu, sd, lower, upper):
    x = float(x)
    mu = float(mu)
    sd = float(sd)
    lower = float(lower)
    upper = float(upper)

    if not (lower <= x <= upper):
        return -np.inf

    a = (lower - mu) / sd
    b = (upper - mu) / sd
    Z = norm.cdf(b) - norm.cdf(a)
    return float(-0.5 * ((x - mu) / sd) ** 2 - np.log(sd * np.sqrt(2 * np.pi) * Z))


def truncnorm_density_1d(x, mu, sd, lower, upper):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    a = (float(lower) - float(mu)) / float(sd)
    b = (float(upper) - float(mu)) / float(sd)
    Z = norm.cdf(b) - norm.cdf(a)
    mask = (x >= float(lower)) & (x <= float(upper))
    z = (x[mask] - float(mu)) / float(sd)
    out[mask] = np.exp(-0.5 * z * z) / (float(sd) * np.sqrt(2 * np.pi) * Z)
    return out


def lognormal_density_1d(x, phi_mu, phi_sd):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    mask = x > 0
    z = (np.log(x[mask]) - float(phi_mu)) / float(phi_sd)
    out[mask] = np.exp(-0.5 * z * z) / (x[mask] * float(phi_sd) * np.sqrt(2 * np.pi))
    return out


# --- lognormal prior for comparison, with original-scale mean equal to the true value ---
K12_TRUE = 5.88
K12_LOGNORMAL_SD = 0.5
# For LogNormal(mu_log, sigma_log^2), mean = exp(mu_log + sigma_log^2 / 2).
# Therefore this choice makes E[k12] = K12_TRUE.
phi_mu_old_k12 = np.array([np.log(K12_TRUE) - 0.5 * K12_LOGNORMAL_SD**2], dtype=float)
phi_sd_old_k12 = np.array([K12_LOGNORMAL_SD], dtype=float)

# --- active truncated normal prior on the original k12 scale ---
# Centered at the true value, but deliberately wide.
K12_center = K12_TRUE
K12_MIN = 0.05
K12_MAX = 12.0
phi_mu = np.array([K12_center], dtype=float)   # parameter-space mean/center
phi_sd = np.array([3.0], dtype=float)          # wide / lower-peak prior

print('Using truncated normal prior for k12')
print(f'true value     = {K12_TRUE}')
print(f'prior mean     = {phi_mu[0]}')
print(f'prior sd       = {phi_sd[0]}')
print(f'bounds         = [{K12_MIN}, {K12_MAX}]')
print(f'comparison lognormal mean = {np.exp(phi_mu_old_k12[0] + 0.5 * phi_sd_old_k12[0]**2):.4f}')

# --- small comparison plot: lognormal mean=true vs truncated normal mean=true ---
xgrid_k12_prior = np.linspace(K12_MIN, K12_MAX, 600)
lognormal_k12 = lognormal_density_1d(xgrid_k12_prior, phi_mu_old_k12[0], phi_sd_old_k12[0])
normal_k12 = truncnorm_density_1d(xgrid_k12_prior, phi_mu[0], phi_sd[0], K12_MIN, K12_MAX)

plt.figure(figsize=(6, 4))
plt.plot(xgrid_k12_prior, lognormal_k12, label='lognormal prior, mean=true')
plt.plot(xgrid_k12_prior, normal_k12, label='truncated normal prior, mean=true')
plt.axvline(K12_TRUE, linestyle='--', label='true k12')
plt.axvline(K12_center, linestyle=':', label='prior mean / center')
plt.xlabel('k12')
plt.ylabel('density')
plt.title('k12 prior: mean at true value')
plt.legend()
plt.tight_layout()
plt.show()


# --- patch SMC-ABC to use the truncated normal prior directly on k12 ---
def sample_prior_phi_k12(rng, phi_mu, phi_sd, n):
    vals = truncnorm_sample_scalar(rng, phi_mu[0], phi_sd[0], K12_MIN, K12_MAX, int(n))
    return vals.reshape(-1, 1)


def log_prior_phi_k12(phi, phi_mu, phi_sd):
    phi = np.asarray(phi, dtype=float)
    return truncnorm_logpdf_scalar(phi[0], phi_mu[0], phi_sd[0], K12_MIN, K12_MAX)


def evaluate_candidate_phi_k12(
    phi,
    times_obs,
    summary_obs,
    n_reps,
    base_seed,
    theta_transform=None,
    sim_kwargs=None,
):
    """
    One-dimensional ABC evaluator for k12.
    Here phi[0] is k12 itself, not log(k12).
    """
    phi = np.asarray(phi, dtype=float)
    k12 = float(phi[0])

    if not np.isfinite(k12):
        return phi, np.inf
    if not (K12_MIN <= k12 <= K12_MAX):
        return phi, np.inf

    paths = []
    rng = np.random.default_rng(int(base_seed))

    try:
        for _ in range(int(n_reps)):
            seed = int(rng.integers(0, 2**31 - 1))
            paths.append(simulate_one_k12_attach_only(np.array([k12], dtype=float), times_obs, seed=seed))

        paths = np.asarray(paths, dtype=float)
        summary_sim = smc_abc.summarize_paths(paths)
        d = smc_abc.l2_distance(summary_sim, summary_obs)

        if not np.isfinite(d):
            return phi, np.inf

        return phi, float(d)
    except Exception:
        return phi, np.inf


smc_abc.simulate_one = simulate_one_k12_attach_only
smc_abc.sample_prior_phi = sample_prior_phi_k12
smc_abc.log_prior_phi = log_prior_phi_k12
smc_abc.evaluate_candidate_phi = evaluate_candidate_phi_k12

print('Patched smc_abc for k12 with a true-centered truncated normal prior in parameter space.')
```

## 8. Run SMC-ABC for $k_{12}$ with the truncated normal prior

The SMC internals still use the variable name `phi`, but in this section `phi[0] = k12` directly.


```python
res = smc_abc.smc_abc_prc1(
    times_obs=times_obs,
    y_obs=y_obs,
    phi_mu=phi_mu,
    phi_sd=phi_sd,
    P=30,
    G=3,
    pool=100,
    n_reps=5,
    eps_quantile=50.0,
    cov_scale=2.0,
    seed=1,
    n_jobs=-1,
    batch_factor=4,
)

print("Finished SMC-ABC.")
print("eps history:", res.eps_history)
```

## 9. Read posterior particles for $k_{12}$

Since this section now uses a direct parameter-space prior, the stored particles are already on the $k_{12}$ scale.


```python
posterior_k12 = np.asarray(res.particles_phi[:, 0], dtype=float)
posterior_w = np.asarray(res.weights, dtype=float)

print("posterior k12 particles:")
print(posterior_k12)
print("posterior weights:")
print(posterior_w)
```

## 10. Posterior summaries

```python
post_mean = np.sum(posterior_w * posterior_k12)

order = np.argsort(posterior_k12)
k12_sorted = posterior_k12[order]
w_sorted = posterior_w[order]
cdf = np.cumsum(w_sorted)
post_median = k12_sorted[np.searchsorted(cdf, 0.5)]

print("True k12:", K12_TRUE)
print("Posterior mean:", post_mean)
print("Posterior median:", post_median)
```

## 11. Posterior plot

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.hist(posterior_k12, bins=10, weights=posterior_w)
plt.axvline(K12_center, linestyle=":", label="prior center")
plt.axvline(K12_TRUE, linestyle="--", label="true value")
plt.xlabel("k12")
plt.ylabel("posterior weight")
plt.title("Posterior for k12 with attachment-only cooperativity")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

posterior_samples = np.asarray(posterior_k12, dtype=float)
posterior_w = np.asarray(posterior_w, dtype=float)

kde_post = gaussian_kde(posterior_samples, weights=posterior_w)

xmin = K12_MIN
xmax = K12_MAX
xgrid = np.linspace(xmin, xmax, 500)

prior_y = truncnorm_density_1d(xgrid, phi_mu[0], phi_sd[0], K12_MIN, K12_MAX)
post_y = kde_post(xgrid)

plt.figure(figsize=(6, 4))
plt.plot(xgrid, prior_y, label='Prior')
plt.plot(xgrid, post_y, label='Posterior')
plt.axvline(K12_center, linestyle=':', label='Prior mean')
plt.axvline(K12_TRUE, linestyle='--', label='True value')
plt.xlabel('k12')
plt.ylabel('Density')
plt.title('Prior vs posterior for k12 (truncated normal prior)')
plt.legend()
plt.tight_layout()
plt.show()
```

## 18. Patch SMC-ABC for one-dimensional $k_{21}$ inference

Here `phi[0] = k21` directly, and we patch the prior sampler and prior density to use the truncated normal prior.


```python
def sample_prior_phi_k21(rng, phi_mu, phi_sd, n):
    vals = truncnorm_sample_scalar(rng, phi_mu[0], phi_sd[0], K21_MIN, K21_MAX, int(n))
    return vals.reshape(-1, 1)


def log_prior_phi_k21(phi, phi_mu, phi_sd):
    phi = np.asarray(phi, dtype=float)
    return truncnorm_logpdf_scalar(phi[0], phi_mu[0], phi_sd[0], K21_MIN, K21_MAX)


def evaluate_candidate_phi_k21(
    phi,
    times_obs,
    summary_obs,
    n_reps,
    base_seed,
    theta_transform=None,
    sim_kwargs=None,
):
    """
    One-dimensional ABC evaluator for k21.
    Here phi[0] is k21 itself, not log(k21).
    """
    phi = np.asarray(phi, dtype=float)
    k21 = float(phi[0])

    if not np.isfinite(k21):
        return phi, np.inf
    if not (K21_MIN <= k21 <= K21_MAX):
        return phi, np.inf

    paths = []
    rng = np.random.default_rng(int(base_seed))

    try:
        for _ in range(int(n_reps)):
            seed = int(rng.integers(0, 2**31 - 1))
            paths.append(
                simulate_one_k21_detach_only(
                    np.array([k21], dtype=float),
                    times_obs,
                    seed=seed,
                )
            )

        paths = np.asarray(paths, dtype=float)
        summary_sim = smc_abc.summarize_paths(paths)
        d = smc_abc.l2_distance(summary_sim, summary_obs)

        if not np.isfinite(d):
            return phi, np.inf

        return phi, float(d)

    except Exception:
        return phi, np.inf


smc_abc.simulate_one = simulate_one_k21_detach_only
smc_abc.sample_prior_phi = sample_prior_phi_k21
smc_abc.log_prior_phi = log_prior_phi_k21
smc_abc.evaluate_candidate_phi = evaluate_candidate_phi_k21

print('Patched smc_abc for one-parameter k21 inference with a truncated normal prior.')
```

## 19. Run SMC-ABC for $k_{21}$

Start with small values while debugging. Increase `P`, `G`, and `pool` for a stronger run.


```python

res_k21 = smc_abc.smc_abc_prc1(
    times_obs=times_obs,
    y_obs=y_obs_k21,
    phi_mu=phi_mu_k21,
    phi_sd=phi_sd_k21,
    P=30,
    G=3,
    pool=100,
    n_reps=5,
    eps_quantile=50.0,
    cov_scale=2.0,
    seed=2,
    n_jobs=-1,
    batch_factor=4,
)

print("Finished SMC-ABC for k21.")
print("eps history:", res_k21.eps_history)
```


## 20. Posterior summaries for $k_{21}$


```python
posterior_k21 = np.asarray(res_k21.particles_phi[:, 0], dtype=float)
posterior_w_k21 = np.asarray(res_k21.weights, dtype=float)

post_mean_k21 = float(np.sum(posterior_w_k21 * posterior_k21))

order = np.argsort(posterior_k21)
k21_sorted = posterior_k21[order]
w_sorted = posterior_w_k21[order]
cdf = np.cumsum(w_sorted)
post_median_k21 = float(k21_sorted[np.searchsorted(cdf, 0.5)])

print("posterior k21 particles:")
print(posterior_k21)
print("posterior weights:")
print(posterior_w_k21)
print()
print("True k21:", K21_TRUE)
print("Prior mean:", phi_mu_k21[0])
print("Posterior mean:", post_mean_k21)
print("Posterior median:", post_median_k21)
```

## 21. Prior vs posterior plot for $k_{21}$
