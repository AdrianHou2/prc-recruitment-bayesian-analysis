# One-parameter SMC-ABC for k12 with hopping turned off

This notebook performs a **one-parameter inference** for

- `k12` = `base_double_attachment_rate`

using the spatial PRC1 Gillespie model, with:

- all other kinetic parameters fixed,
- a fixed cooperativity energy,
- `enable_cooperativity=True`,
- `enable_hopping=False`.

Before running SMC-ABC, it performs **one timing check** for a single simulation.

## 1. Load local modules

```python
from pathlib import Path
import importlib.util
import inspect
import sys

def load_module(module_name, filename_candidates):
    search_roots = [
        Path.cwd(),
        Path.cwd() / "python_version",
        Path.cwd().parent,
    ]
    for root in search_roots:
        for name in filename_candidates:
            path = root / name
            if path.exists():
                spec = importlib.util.spec_from_file_location(module_name, path)
                module = importlib.util.module_from_spec(spec)
                sys.modules[module_name] = module
                spec.loader.exec_module(module)
                print(f"Loaded {module_name} from: {path}")
                return module
    raise FileNotFoundError(f"Could not find any of: {filename_candidates}")

gillespie = load_module("gillespie", ["gillespie.py", "gillespie(20).py", "gillespie(7).py"])
prc1 = load_module("prc1", ["prc1.py", "prc1(19).py", "prc1(7).py"])
prc1_state = load_module("prc1_state", ["prc1_state.py", "prc1_state(18).py", "prc1_state(7).py"])
run_gillespie = load_module("run_gillespie", ["run_gillespie.py", "run_gillespie(18).py", "run_gillespie(6).py"])
smc_abc = load_module("smc_abc", ["smc_abc.py", "smc_abc(6).py", "smc-abc.py", "smc-abc(6).py"])

sig = inspect.signature(run_gillespie.run_gillespie_prc1_on_grid)
required_flags = {
    "enable_initial_attach_cooperativity",
    "enable_second_head_attach_cooperativity",
    "enable_second_head_detach_cooperativity",
}
missing = sorted(required_flags - set(sig.parameters))
if missing:
    raise RuntimeError(
        "Your loaded run_gillespie module does not have the new cooperativity flags. "
        f"Missing: {missing}. Make sure you are loading the patched files."
    )

print("Patched cooperativity flags detected successfully.")
```

## 2. Fixed parameters and observation grid

```python
import numpy as np

# fixed model parameters
INIT_BIND = 0.002
K_DET_1 = 0.1
K_DET_2 = 0.1
MAX_STEPS = 200_000

# observation grid
times_obs = np.arange(0.0, 68.0 + 4.0, 4.0)

# thermal scale and fixed cooperativity
k_B_T = 4.1
EPS_FIXED = 2.3 * k_B_T

# true k12 used for synthetic data
K12_TRUE = 5.88

print("times_obs =", times_obs)
print("EPS_FIXED =", EPS_FIXED)
print("K_DET_1 =", K_DET_1)
print("K_DET_2 =", K_DET_2)
print("K12_TRUE =", K12_TRUE)
```

## 3. Define a one-parameter simulator

Here `theta[0]` is `k12 = base_double_attachment_rate`.
Everything else is fixed, and hopping is disabled.

```python
def simulate_one_k12_attach_only(theta, times_obs, seed):
    """
    theta[0] = k12 = base_double_attachment_rate
    All other parameters are fixed.
    Hopping is disabled.
    Cooperativity acts only in second-head attachment.
    """
    import numpy as np
    np.random.seed(int(seed))

    return run_gillespie.run_gillespie_prc1_on_grid(
        initial_binding_rate_per_site=INIT_BIND,
        singly_bound_detachment_rate=K_DET_1,
        base_double_attachment_rate=float(theta[0]),
        base_double_detachment_rate=K_DET_2,
        times_obs=times_obs,
        cooperativity_energy=EPS_FIXED,
        enable_initial_attach_cooperativity=False,
        enable_second_head_attach_cooperativity=True,
        enable_second_head_detach_cooperativity=False,
        enable_hopping=False,
        max_steps=MAX_STEPS,
    )
```

## 4. Time one simulation before doing inference

```python
import time

theta_test = np.array([K12_TRUE], dtype=float)

t0 = time.perf_counter()
test_path = simulate_one_k12_attach_only(theta_test, times_obs, seed=123)
t1 = time.perf_counter()

one_sim_seconds = t1 - t0

print("One simulation path:")
print(test_path)
print("Shape:", test_path.shape)
print(f"Time for one simulation: {one_sim_seconds:.6f} seconds")
```

## 5. Patch the SMC-ABC module to use the one-parameter simulator

```python
smc_abc.simulate_one = simulate_one_k12_attach_only
print("Patched smc_abc.simulate_one successfully.")
print("joblib available:", getattr(smc_abc, "_JOBLIB_AVAILABLE", None))
```

## 6. Generate synthetic observed data

```python
n_obs_reps = 10

y_obs = np.array([
    run_gillespie.run_gillespie_prc1_on_grid(
        initial_binding_rate_per_site=INIT_BIND,
        singly_bound_detachment_rate=K_DET_1,
        base_double_attachment_rate=K12_TRUE,
        base_double_detachment_rate=K_DET_2,
        times_obs=times_obs,
        cooperativity_energy=EPS_FIXED,
        enable_initial_attach_cooperativity=False,
        enable_second_head_attach_cooperativity=True,
        enable_second_head_detach_cooperativity=False,
        enable_hopping=False,
        max_steps=MAX_STEPS,
    )
    for _ in range(n_obs_reps)
], dtype=float)

print("y_obs shape:", y_obs.shape)
print(y_obs)
```

## 7. Set a one-parameter prior in log-space

```python
K12_TRUE = 5.88
K12_center = 2.0
phi_mu = np.array([np.log(K12_center)], dtype=float)
phi_sd = np.array([0.5], dtype=float)
```

## 8. Run SMC-ABC

Set `n_jobs=1` first while debugging.
If the code is stable and your `smc_abc` parallel backend is working, you can change it later.

```python
res = smc_abc.smc_abc_prc1(
    times_obs=times_obs,
    y_obs=y_obs,
    phi_mu=phi_mu,
    phi_sd=phi_sd,
    P=30,
    G=3,
    pool=100,
    n_reps=5,
    eps_quantile=50.0,
    cov_scale=2.0,
    seed=1,
    n_jobs=-1,
    batch_factor=4,
)

print("Finished SMC-ABC.")
print("eps history:", res.eps_history)
```

## 9. Convert posterior particles back to k12 scale

```python
posterior_k12 = np.exp(res.particles_phi[:, 0])
posterior_w = res.weights

print("posterior k12 particles:")
print(posterior_k12)
print("posterior weights:")
print(posterior_w)
```

## 10. Posterior summaries

```python
post_mean = np.sum(posterior_w * posterior_k12)

order = np.argsort(posterior_k12)
k12_sorted = posterior_k12[order]
w_sorted = posterior_w[order]
cdf = np.cumsum(w_sorted)
post_median = k12_sorted[np.searchsorted(cdf, 0.5)]

print("True k12:", K12_TRUE)
print("Posterior mean:", post_mean)
print("Posterior median:", post_median)
```

## 11. Posterior plot

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.hist(posterior_k12, bins=10, weights=posterior_w)
plt.axvline(K12_center, linestyle=":", label="prior center")
plt.axvline(K12_TRUE, linestyle="--", label="true value")
plt.xlabel("k12")
plt.ylabel("posterior weight")
plt.title("Posterior for k12 with attachment-only cooperativity")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

posterior_samples = np.asarray(posterior_k12, dtype=float)
posterior_w = np.asarray(posterior_w, dtype=float)

phi0 = float(phi_mu[0])
sd0 = float(phi_sd[0])

def prior_density(x):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    mask = x > 0
    z = (np.log(x[mask]) - phi0) / sd0
    out[mask] = np.exp(-0.5 * z * z) / (x[mask] * sd0 * np.sqrt(2 * np.pi))
    return out

kde_post = gaussian_kde(posterior_samples, weights=posterior_w)

xmin = min(np.min(posterior_samples), K12_TRUE, K12_center) * 0.6
xmax = max(np.max(posterior_samples), K12_TRUE, K12_center) * 1.4
xmin = max(xmin, 1e-8)

xgrid = np.linspace(xmin, xmax, 500)

prior_y = prior_density(xgrid)
post_y = kde_post(xgrid)

plt.figure(figsize=(6, 4))
plt.plot(xgrid, prior_y, label="Prior")
plt.plot(xgrid, post_y, label="Posterior")
plt.axvline(K12_center, linestyle=":", label="Prior center")
plt.axvline(K12_TRUE, linestyle="--", label="True value")
plt.xlabel("k12")
plt.ylabel("Density")
plt.title("Prior vs posterior for k12 (inaccurate prior)")
plt.legend()
plt.show()
```

# 12. One-parameter SMC-ABC for $k_{21}$ with detachment cooperativity only

This section adds the matching one-parameter inference for

\[
k_{21}=\texttt{base\_double\_detachment\_rate}.
\]

Here the only cooperativity effect turned on is **second-head detachment cooperativity**:

\[
\texttt{enable\_second\_head\_detach\_cooperativity=True}.
\]

Second-head attachment cooperativity is turned off. Hopping is also turned off.


```python

# ── k21 inference setup ──────────────────────────────────────────────────────
# We reuse the fixed values from the k12 section:
#   INIT_BIND, K_DET_1, EPS_FIXED, K12_TRUE, MAX_STEPS, times_obs

K12_FIXED_FOR_K21 = K12_TRUE
K21_TRUE = K_DET_2       # paper-like value: 0.1

print("K21 inference setup")
print("INIT_BIND          =", INIT_BIND)
print("K_DET_1            =", K_DET_1)
print("K12_FIXED_FOR_K21  =", K12_FIXED_FOR_K21)
print("K21_TRUE           =", K21_TRUE)
print("EPS_FIXED          =", EPS_FIXED)
```


## 13. Define the $k_{21}$ simulator

This simulator fixes everything except `base_double_detachment_rate`.  It uses detachment cooperativity only.


```python

def simulate_one_k21_detach_only(theta, times_obs, seed):
    """
    theta[0] = k21 = base_double_detachment_rate.

    All other parameters are fixed.
    Hopping is disabled.
    Cooperativity acts only in second-head detachment.
    """
    import numpy as np
    np.random.seed(int(seed))

    return run_gillespie.run_gillespie_prc1_on_grid(
        initial_binding_rate_per_site=INIT_BIND,
        singly_bound_detachment_rate=K_DET_1,
        base_double_attachment_rate=K12_FIXED_FOR_K21,
        base_double_detachment_rate=float(theta[0]),
        times_obs=times_obs,
        cooperativity_energy=EPS_FIXED,
        enable_initial_attach_cooperativity=False,
        enable_second_head_attach_cooperativity=False,
        enable_second_head_detach_cooperativity=True,
        enable_hopping=False,
        max_steps=MAX_STEPS,
    )
```


## 14. Time one $k_{21}$ simulation


```python

import time

theta_test_k21 = np.array([K21_TRUE], dtype=float)

t0 = time.perf_counter()
test_path_k21 = simulate_one_k21_detach_only(theta_test_k21, times_obs, seed=456)
t1 = time.perf_counter()

one_sim_seconds_k21 = t1 - t0

print("One k21 simulation path:")
print(test_path_k21)
print("Shape:", test_path_k21.shape)
print(f"Time for one k21 simulation: {one_sim_seconds_k21:.6f} seconds")
```


## 15. Generate synthetic observed data for $k_{21}$

The observed data are generated at the true value $k_{21}=0.1$ with detachment cooperativity only.


```python

n_obs_reps_k21 = 10

y_obs_k21 = []
for r in range(n_obs_reps_k21):
    y_obs_k21.append(
        simulate_one_k21_detach_only(
            np.array([K21_TRUE], dtype=float),
            times_obs,
            seed=10_000 + r,
        )
    )

y_obs_k21 = np.asarray(y_obs_k21, dtype=float)

print("y_obs_k21 shape:", y_obs_k21.shape)
print(y_obs_k21)
print("mean y_obs_k21:")
print(y_obs_k21.mean(axis=0))
```


## 16. Wider/lower-peak lognormal prior for $k_{21}$

For the lognormal prior, increasing `phi_sd` from 0.5 to around 1.0 makes the prior more spread out and lowers the main peak compared with the tighter prior, without going to the unstable `phi_sd = 2.0` case.

Here the prior median is intentionally centered away from the truth:

\[
\text{prior median}=0.3,
\qquad
k_{21}^{\star}=0.1.
\]


```python

# ── Prior in log-space for k21 ───────────────────────────────────────────────
# phi = log(k21)
#
# To lower the peak compared with phi_sd=0.5, use a wider log-space sd.
# Avoid phi_sd=2.0 because it creates a very long tail and a near-zero spike.

K21_center = 0.3
phi_mu_k21 = np.array([np.log(K21_center)], dtype=float)
phi_sd_k21 = np.array([1.0], dtype=float)   # wider/lower-peak prior than 0.5

# Physical/numerical bounds for rejecting proposals.
# These bounds do not change the prior formula used by smc_abc.log_prior_phi,
# but they prevent wasting simulation time on unreasonable proposals.
K21_MIN = 0.01
K21_MAX = 1.0

print(f"Inferred parameter: k21")
print(f"True value    : k21 = {K21_TRUE}")
print(f"Prior median  : k21 = {K21_center}")
print(f"phi_mu_k21    = {phi_mu_k21.round(4)}")
print(f"phi_sd_k21    = {phi_sd_k21}")
print(f"bounds        = [{K21_MIN}, {K21_MAX}]")
print(f"lognormal mode = exp(mu - sd^2) = {np.exp(phi_mu_k21[0] - phi_sd_k21[0]**2):.5g}")
```


## 17. Plot the prior shape before running inference

This plot compares the old tighter prior width `phi_sd=0.5` with the wider prior used here.


```python

import numpy as np
import matplotlib.pyplot as plt

def lognormal_density_1d(x, phi_mu, phi_sd):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    mask = x > 0
    z = (np.log(x[mask]) - float(phi_mu)) / float(phi_sd)
    out[mask] = np.exp(-0.5 * z * z) / (x[mask] * float(phi_sd) * np.sqrt(2 * np.pi))
    return out

xgrid_prior_k21 = np.linspace(K21_MIN, K21_MAX, 600)

prior_tight = lognormal_density_1d(xgrid_prior_k21, phi_mu_k21[0], 0.5)
prior_wide = lognormal_density_1d(xgrid_prior_k21, phi_mu_k21[0], phi_sd_k21[0])

plt.figure(figsize=(6, 4))
plt.plot(xgrid_prior_k21, prior_tight, label="old tighter prior, phi_sd=0.5")
plt.plot(xgrid_prior_k21, prior_wide, label=f"used wider prior, phi_sd={phi_sd_k21[0]}")
plt.axvline(K21_TRUE, linestyle="--", label="true k21")
plt.axvline(K21_center, linestyle=":", label="prior median")
plt.xlabel("k21")
plt.ylabel("density")
plt.title("Prior shape for k21")
plt.legend()
plt.tight_layout()
plt.show()
```


## 18. Patch SMC-ABC for one-dimensional $k_{21}$ inference

This patch avoids the default four-parameter transform in `smc_abc.py`.  It also rejects proposals outside the chosen physical bounds.


```python

def evaluate_candidate_phi_k21(
    phi,
    times_obs,
    summary_obs,
    n_reps,
    base_seed,
    theta_transform=None,
    sim_kwargs=None,
):
    """
    One-dimensional ABC evaluator for k21.

    phi[0] = log(k21)
    """
    phi = np.asarray(phi, dtype=float)
    theta_raw = np.exp(phi)
    k21 = float(theta_raw[0])

    # Reject nonfinite or unreasonable proposals.
    if not np.isfinite(k21):
        return phi, np.inf
    if not (K21_MIN <= k21 <= K21_MAX):
        return phi, np.inf

    paths = []
    rng = np.random.default_rng(int(base_seed))

    try:
        for _ in range(int(n_reps)):
            seed = int(rng.integers(0, 2**31 - 1))
            paths.append(
                simulate_one_k21_detach_only(
                    np.array([k21], dtype=float),
                    times_obs,
                    seed=seed,
                )
            )

        paths = np.asarray(paths, dtype=float)
        summary_sim = smc_abc.summarize_paths(paths)
        d = smc_abc.l2_distance(summary_sim, summary_obs)

        if not np.isfinite(d):
            return phi, np.inf

        return phi, float(d)

    except Exception:
        # Failed simulations are rejected by ABC rather than crashing the full SMC run.
        return phi, np.inf

# Patch smc_abc for this one-parameter section.
smc_abc.simulate_one = simulate_one_k21_detach_only
smc_abc.evaluate_candidate_phi = evaluate_candidate_phi_k21

print("Patched smc_abc for one-parameter k21 inference.")
```


## 19. Run SMC-ABC for $k_{21}$

Start with small values while debugging. Increase `P`, `G`, and `pool` for a stronger run.


```python

res_k21 = smc_abc.smc_abc_prc1(
    times_obs=times_obs,
    y_obs=y_obs_k21,
    phi_mu=phi_mu_k21,
    phi_sd=phi_sd_k21,
    P=30,
    G=3,
    pool=100,
    n_reps=5,
    eps_quantile=50.0,
    cov_scale=2.0,
    seed=2,
    n_jobs=-1,
    batch_factor=4,
)

print("Finished SMC-ABC for k21.")
print("eps history:", res_k21.eps_history)
```


## 20. Posterior summaries for $k_{21}$


```python

posterior_k21 = np.exp(res_k21.particles_phi[:, 0])
posterior_w_k21 = np.asarray(res_k21.weights, dtype=float)

post_mean_k21 = float(np.sum(posterior_w_k21 * posterior_k21))

order = np.argsort(posterior_k21)
k21_sorted = posterior_k21[order]
w_sorted = posterior_w_k21[order]
cdf = np.cumsum(w_sorted)
post_median_k21 = float(k21_sorted[np.searchsorted(cdf, 0.5)])

print("posterior k21 particles:")
print(posterior_k21)
print("posterior weights:")
print(posterior_w_k21)
print()
print("True k21:", K21_TRUE)
print("Prior median:", K21_center)
print("Posterior mean:", post_mean_k21)
print("Posterior median:", post_median_k21)
```


## 21. Prior vs posterior plot for $k_{21}$


```python

from scipy.stats import gaussian_kde

posterior_samples_k21 = np.asarray(posterior_k21, dtype=float)
posterior_w_k21 = np.asarray(posterior_w_k21, dtype=float)

kde_post_k21 = gaussian_kde(posterior_samples_k21, weights=posterior_w_k21)

xmin = min(np.min(posterior_samples_k21), K21_TRUE, K21_center, K21_MIN) * 0.8
xmax = max(np.max(posterior_samples_k21), K21_TRUE, K21_center, K21_MAX) * 1.05
xmin = max(xmin, 1e-8)

xgrid_k21 = np.linspace(xmin, xmax, 600)

prior_y_k21 = lognormal_density_1d(xgrid_k21, phi_mu_k21[0], phi_sd_k21[0])
post_y_k21 = kde_post_k21(xgrid_k21)

plt.figure(figsize=(6, 4))
plt.plot(xgrid_k21, prior_y_k21, label="Prior")
plt.plot(xgrid_k21, post_y_k21, label="Posterior")
plt.axvline(K21_center, linestyle=":", label="Prior median")
plt.axvline(K21_TRUE, linestyle="--", label="True value")
plt.xlabel("k21")
plt.ylabel("Density")
plt.title("Prior vs posterior for k21, detachment cooperativity only")
plt.legend()
plt.tight_layout()
plt.show()
```
