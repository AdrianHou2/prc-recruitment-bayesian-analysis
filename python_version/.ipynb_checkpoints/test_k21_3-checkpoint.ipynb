# SMC-ABC: Joint inference of `k21` and `cooperativity_energy`

**Setup:**
- Cooperativity is enabled **only** for second-head detachment (`enable_second_head_detach_cooperativity=True`).
- Inferred parameters: `k21` (double detachment rate) and `cooperativity_energy` (ε).
- All other parameters are fixed.
- Synthetic observed data is generated at the known true values.

```python
from pathlib import Path
import importlib.util
import inspect
import sys

def load_module(module_name, filename_candidates):
    search_roots = [
        Path.cwd(),
        Path.cwd() / "python_version",
        Path.cwd().parent,
    ]
    for root in search_roots:
        for name in filename_candidates:
            path = root / name
            if path.exists():
                spec = importlib.util.spec_from_file_location(module_name, path)
                module = importlib.util.module_from_spec(spec)
                sys.modules[module_name] = module
                spec.loader.exec_module(module)
                print(f"Loaded {module_name} from: {path}")
                return module
    raise FileNotFoundError(f"Could not find any of: {filename_candidates}")

gillespie   = load_module("gillespie",    ["gillespie.py",    "gillespie(7).py"])
prc1        = load_module("prc1",         ["prc1.py",         "prc1(7).py"])
prc1_state  = load_module("prc1_state",   ["prc1_state.py",   "prc1_state(7).py"])
run_gillespie = load_module("run_gillespie", ["run_gillespie.py", "run_gillespie(6).py"])
smc_abc     = load_module("smc_abc",      ["smc_abc.py",      "smc-abc.py", "smc-abc(6).py"])

# Sanity check: confirm cooperativity flags are present
sig = inspect.signature(run_gillespie.run_gillespie_prc1_on_grid)
required_flags = {
    "enable_initial_attach_cooperativity",
    "enable_second_head_attach_cooperativity",
    "enable_second_head_detach_cooperativity",
}
missing = sorted(required_flags - set(sig.parameters))
if missing:
    raise RuntimeError(
        f"run_gillespie is missing cooperativity flags: {missing}. Load the patched files."
    )
print("Cooperativity flags detected successfully.")
```

```python
import numpy as np

# ── Fixed kinetic parameters (not inferred) ──────────────────────────────────
INIT_BIND   = 0.002   # initial_binding_rate_per_site  (R01)
K_DET_1     = 0.1     # singly_bound_detachment_rate   (k10)
K12_FIXED   = 5.88    # base_double_attachment_rate    (k12)
MAX_STEPS   = 50_000

# ── True values of the two inferred parameters ───────────────────────────────
K21_TRUE    = 0.1                  # base_double_detachment_rate
k_B_T       = 4.1                  # thermal energy [pN·nm]
EPS_TRUE    = 2.3 * k_B_T          # cooperativity_energy  ≈ 9.43 pN·nm

# ── Observation grid ─────────────────────────────────────────────────────────
times_obs = np.arange(0.0, 68.0 + 4.0, 4.0)   # 0, 4, 8, …, 68  (18 points)

print(f"K21_TRUE  = {K21_TRUE}")
print(f"EPS_TRUE  = {EPS_TRUE:.4f}  ({EPS_TRUE/k_B_T:.2f} k_BT)")
print(f"times_obs = {times_obs}")
```

## Simulator configuration

Only **detachment cooperativity** is active:
- `enable_initial_attach_cooperativity   = False`
- `enable_second_head_attach_cooperativity = False`
- `enable_second_head_detach_cooperativity = True`   ← only this one

We infer `phi = [log(k21), log(eps)]`.  
Because `cooperativity_energy` is a `sim_kwarg` (not part of `theta_full`), we  
replace `smc_abc.evaluate_candidate_phi` with a 2-D version that builds
`sim_kwargs` dynamically from the current particle.

```python
from numpy.random import default_rng

# Base sim_kwargs – cooperativity_energy will be filled in per-particle
BASE_SIM_KWARGS = dict(
    enable_initial_attach_cooperativity=False,
    enable_second_head_attach_cooperativity=False,
    enable_second_head_detach_cooperativity=True,
    enable_hopping=False,
    max_steps=MAX_STEPS,
)


def theta_transform_2d(theta_raw):
    """
    theta_raw = [k21, eps]  (both in original, positive scale)
    Returns theta_full = [R01, k10, k12, k21].
    eps is handled separately via sim_kwargs.
    """
    theta_raw = np.asarray(theta_raw, dtype=float)
    k21 = float(theta_raw[0])
    return np.array([INIT_BIND, K_DET_1, K12_FIXED, k21], dtype=float)


def evaluate_candidate_phi_2d(
    phi, times_obs, summary_obs, n_reps, base_seed,
    theta_transform=None,   # kept for API compatibility, ignored here
    sim_kwargs=None,         # kept for API compatibility, ignored here
):
    """
    2-D evaluator.
    phi[0] = log(k21),  phi[1] = log(cooperativity_energy)

    Builds sim_kwargs on the fly so cooperativity_energy varies per particle.
    """
    phi = np.asarray(phi, dtype=float)
    theta_raw = np.exp(phi)           # [k21, eps] in original scale

    k21 = float(theta_raw[0])
    eps = float(theta_raw[1])

    theta_full = np.array([INIT_BIND, K_DET_1, K12_FIXED, k21], dtype=float)

    current_sim_kwargs = dict(BASE_SIM_KWARGS)
    current_sim_kwargs["cooperativity_energy"] = eps

    paths = []
    rng = default_rng(int(base_seed))
    for _ in range(int(n_reps)):
        seed = int(rng.integers(0, 2**31 - 1))
        paths.append(
            smc_abc.simulate_one(theta_full, times_obs, seed,
                                 sim_kwargs=current_sim_kwargs)
        )

    paths = np.asarray(paths, dtype=float)
    summary_sim = smc_abc.summarize_paths(paths)
    d = smc_abc.l2_distance(summary_sim, summary_obs)
    return np.asarray(phi, dtype=float), float(d)


# ── Monkey-patch so smc_abc_prc1 / pilot_run use our 2-D evaluator ───────────
smc_abc.evaluate_candidate_phi = evaluate_candidate_phi_2d
print("evaluate_candidate_phi patched to 2-D version.")
print("BASE_SIM_KWARGS =", BASE_SIM_KWARGS)
```

```python
import time

# Quick sanity check: one forward simulation at the true parameters
test_sim_kwargs = dict(BASE_SIM_KWARGS, cooperativity_energy=EPS_TRUE)
theta_true_full = np.array([INIT_BIND, K_DET_1, K12_FIXED, K21_TRUE], dtype=float)

t0 = time.perf_counter()
test_path = smc_abc.simulate_one(theta_true_full, times_obs, seed=42,
                                 sim_kwargs=test_sim_kwargs)
t1 = time.perf_counter()

print("One simulation path at true params:")
print(test_path)
print(f"Wall time: {t1 - t0:.3f} s")
```

```python
# Generate synthetic observed data at the TRUE parameter values
n_obs_reps = 10
true_sim_kwargs = dict(BASE_SIM_KWARGS, cooperativity_energy=EPS_TRUE)

y_obs = np.array([
    smc_abc.simulate_one(
        np.array([INIT_BIND, K_DET_1, K12_FIXED, K21_TRUE], dtype=float),
        times_obs, seed=i, sim_kwargs=true_sim_kwargs
    )
    for i in range(n_obs_reps)
], dtype=float)   # shape (10, 18)

print("y_obs shape:", y_obs.shape)
print("Summary statistic dimension:", smc_abc.summarize_paths(y_obs).shape)
print("Mean PRC1 count at each timepoint:")
print(y_obs.mean(axis=0).round(1))
```

```python
# ── Prior in log-space ────────────────────────────────────────────────────────
# phi = [log(k21), log(eps)]
#
# k21:  true = 0.1,  prior center = 0.3   (intentionally off by 3×)
# eps:  true ≈ 9.43, prior center = 5.0   (intentionally off by ~2×)

K21_center = 0.3
EPS_center = 5.0

phi_mu = np.array([np.log(K21_center), np.log(EPS_center)], dtype=float)
phi_sd = np.array([0.5, 0.5], dtype=float)

print(f"Inferred parameters: k21, cooperativity_energy")
print(f"True values  : k21 = {K21_TRUE},  eps = {EPS_TRUE:.4f}")
print(f"Prior centers: k21 = {K21_center}, eps = {EPS_center}")
print(f"phi_mu = {phi_mu.round(4)}")
print(f"phi_sd = {phi_sd}")
```

```python
# ── Run SMC-ABC ───────────────────────────────────────────────────────────────
# theta_transform is passed but ignored by evaluate_candidate_phi_2d.
# sim_kwargs is also passed for API compatibility but ignored (eps is read from phi).

res = smc_abc.smc_abc_prc1(
    times_obs=times_obs,
    y_obs=y_obs,
    phi_mu=phi_mu,
    phi_sd=phi_sd,
    P=30,
    G=3,
    pool=60,
    n_reps=1,
    eps_quantile=50.0,
    cov_scale=2.0,
    seed=1,
    n_jobs=-1,
    batch_factor=16,
    theta_transform=theta_transform_2d,   # kept for API compat
    sim_kwargs=BASE_SIM_KWARGS,           # kept for API compat
    parallel_backend="threads",
)

print("\nSMC-ABC finished.")
print("Epsilon history:", res.eps_history)
```

```python
# ── Extract posterior particles (back to original scale) ──────────────────────
posterior_k21 = np.exp(res.particles_phi[:, 0])
posterior_eps = np.exp(res.particles_phi[:, 1])
posterior_w   = res.weights

print("posterior k21 particles:", posterior_k21.round(4))
print("posterior eps particles: ", posterior_eps.round(4))
print("weights sum:", posterior_w.sum().round(6))
```

```python
# ── Weighted posterior summaries ──────────────────────────────────────────────
def weighted_median(values, weights):
    order = np.argsort(values)
    cdf = np.cumsum(weights[order])
    return values[order][np.searchsorted(cdf, 0.5)]

k21_mean   = np.sum(posterior_w * posterior_k21)
k21_median = weighted_median(posterior_k21, posterior_w)
eps_mean   = np.sum(posterior_w * posterior_eps)
eps_median = weighted_median(posterior_eps, posterior_w)

print(f"{'':25s} {'k21':>10s}   {'eps':>10s}")
print(f"{'True value':25s} {K21_TRUE:>10.4f}   {EPS_TRUE:>10.4f}")
print(f"{'Prior center':25s} {K21_center:>10.4f}   {EPS_center:>10.4f}")
print(f"{'Posterior mean':25s} {k21_mean:>10.4f}   {eps_mean:>10.4f}")
print(f"{'Posterior median':25s} {k21_median:>10.4f}   {eps_median:>10.4f}")
```

```python
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# ── Helper: log-normal prior density in original scale ────────────────────────
def lognormal_density(x, log_mu, log_sd):
    x = np.asarray(x, dtype=float)
    out = np.zeros_like(x)
    mask = x > 0
    z = (np.log(x[mask]) - log_mu) / log_sd
    out[mask] = np.exp(-0.5 * z * z) / (x[mask] * log_sd * np.sqrt(2 * np.pi))
    return out

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

param_info = [
    dict(samples=posterior_k21, true_val=K21_TRUE, prior_center=K21_center,
         phi_mu=phi_mu[0], phi_sd=phi_sd[0], label="k21",
         title="Posterior for k21 (double detachment rate)"),
    dict(samples=posterior_eps, true_val=EPS_TRUE,  prior_center=EPS_center,
         phi_mu=phi_mu[1], phi_sd=phi_sd[1], label="ε (cooperativity energy)",
         title="Posterior for cooperativity energy ε"),
]

for ax, info in zip(axes, param_info):
    samples = info["samples"]
    w       = posterior_w

    xmin = min(samples.min(), info["true_val"], info["prior_center"]) * 0.5
    xmax = max(samples.max(), info["true_val"], info["prior_center"]) * 1.6
    xmin = max(xmin, 1e-8)
    xgrid = np.linspace(xmin, xmax, 400)

    prior_y = lognormal_density(xgrid, info["phi_mu"], info["phi_sd"])
    kde = gaussian_kde(samples, weights=w)
    post_y = kde(xgrid)

    ax.plot(xgrid, prior_y, label="Prior")
    ax.plot(xgrid, post_y, label="Posterior (KDE)")
    ax.axvline(info["prior_center"], linestyle=":",  color="C0", label="Prior center")
    ax.axvline(info["true_val"],     linestyle="--", color="red", label="True value")
    ax.set_xlabel(info["label"])
    ax.set_ylabel("Density")
    ax.set_title(info["title"])
    ax.legend()

plt.tight_layout()
plt.show()
```

```python
# ── 2-D scatter of posterior particles ───────────────────────────────────────
fig, ax = plt.subplots(figsize=(6, 5))

scatter = ax.scatter(
    posterior_k21, posterior_eps,
    s=posterior_w / posterior_w.max() * 200 + 10,
    c=posterior_w, cmap="viridis", alpha=0.8
)
plt.colorbar(scatter, ax=ax, label="Particle weight")

ax.axvline(K21_TRUE,   linestyle="--", color="red",  label=f"k21 true = {K21_TRUE}")
ax.axhline(EPS_TRUE,   linestyle="--", color="blue", label=f"ε true = {EPS_TRUE:.2f}")
ax.axvline(K21_center, linestyle=":",  color="red",  alpha=0.5, label=f"k21 prior center = {K21_center}")
ax.axhline(EPS_center, linestyle=":",  color="blue", alpha=0.5, label=f"ε prior center = {EPS_center}")

ax.set_xlabel("k21")
ax.set_ylabel("cooperativity energy ε")
ax.set_title("Joint posterior particles\n(detach cooperativity only)")
ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
```
