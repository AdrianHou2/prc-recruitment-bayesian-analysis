
# Spatial PRC1 → saturating population CTMC trajectory check

This is the **saturation-aware** version of the mapping check. It builds on
`spatial_to_ctmc_trajectory_check_actual_k12.ipynb` and changes two things:

1. **The CTMC simulator now uses the saturation factor**
   $\bigl(1 - (n_1 + 2n_2)/n_s\bigr)$
   on attachment propensities, matching the population model in section 1
   of the report. This lets the CTMC bend as the lattice fills.

2. **The mapping extracts intrinsic (zero-occupancy) attachment rates**,
   not time-averaged rates. Otherwise we would double count saturation:
   once via the recorded `p.double_attachment_rate` averaging over a
   filling lattice, and once again via the new $(1-\cdot)$ factor in the CTMC.

Concretely, the new mapping is:
$$
  \widehat{R}_{01} = 2\, k_{01}^{\mathrm{spatial}}\, n_s,
  \qquad
  \widehat{k}_{10} = k_{10}^{\mathrm{spatial}},
$$
$$
  \widehat{k}_{12} = \frac{\sum_{r,t}\sum_i k_{12,i}^{\mathrm{spatial}}(t)/(1 - \rho(t))}
                            {\sum_{r,t} n_1(t)},
  \qquad
  \widehat{k}_{21} = \frac{\sum_{r,t}\sum_j k_{21,j}^{\mathrm{spatial}}(t)}
                            {\sum_{r,t} n_2(t)},
$$
where $\rho(t) = (n_1(t) + 2n_2(t))/n_s$ is the local occupancy at the
snapshot when each rate was recorded, and $n_s = $ number of sites per
microtubule (= 626 with the default geometry, since 5000/8 ≈ 625 plus a
small boundary correction handled by the spatial code).

Detachment rates are unchanged — the population model in §1 doesn't apply
any saturation factor to them.

## Diagnostic definitions

The map used here is:

\[
R_{01}\approx 2 k_{01}^{\mathrm{site}} A^{\mathrm{spatial}},
\qquad
k_{10}^{\mathrm{CTMC}} = k_{10}^{\mathrm{spatial}}.
\]

For \(k_{12}\), instead of using only the rough count
\[
k_{12}\approx k_{12}^{\mathrm{site}}\widehat m,
\]
this notebook records the actual spatial second-head attachment rate for each singly bound PRC1:
\[
k_{12,i}^{\mathrm{spatial}}(t)
=
\texttt{p.double\_attachment\_rate}.
\]
Then it estimates
\[
\widehat k_{12}
=
\frac{
\sum_{r,t}\sum_{i=1}^{n_1(t)}
k_{12,i}^{\mathrm{spatial}}(t)
}{
\sum_{r,t} n_1(t)
}.
\]

For \(k_{21}\), this notebook records the actual double-to-single detachment hazard for each doubly bound PRC1:
\[
k_{21,j}^{\mathrm{spatial}}(t)
=
\texttt{p.detachment\_rate}.
\]
This is the sum of the two possible head-detachment rates and matches the CTMC reaction
\[
n_2\to n_1,
\qquad
a_4(X)=k_{21}n_2.
\]
Thus
\[
\widehat k_{21}
=
\frac{
\sum_{r,t}\sum_{j=1}^{n_2(t)}
k_{21,j}^{\mathrm{spatial}}(t)
}{
\sum_{r,t} n_2(t)
}.
\]

The rough \(\widehat m\) is still recorded only as a diagnostic, not used to set \(k_{12}\) in this version.


```python
def make_spatial_state(
    initial_binding_rate_per_site,
    singly_bound_detachment_rate,
    base_double_attachment_rate,
    base_double_detachment_rate,
    cooperativity_energy=0.0,
    enable_initial_attach_cooperativity=False,
    enable_second_head_attach_cooperativity=False,
    enable_second_head_detach_cooperativity=False,
):
    """
    Same fixed geometry as run_gillespie_prc1 in the uploaded code.
    """
    microtubule_length = 5000
    site_spacing = 8
    microtubule_offset = 0
    spring_constant = 2
    rest_length = 32
    k_B_T = 4.1
    microtubule_separation = 30
    base_hopping_rate = 1

    return State(
        microtubule_length,
        site_spacing,
        microtubule_offset,
        spring_constant,
        rest_length,
        k_B_T,
        microtubule_separation,
        initial_binding_rate_per_site,
        singly_bound_detachment_rate,
        base_double_attachment_rate,
        base_double_detachment_rate,
        base_hopping_rate,
        cooperativity_energy,
        enable_initial_attach_cooperativity,
        enable_second_head_attach_cooperativity,
        enable_second_head_detach_cooperativity,
    )


def count_second_head_targets_rough(prc1):
    """
    Rough count for m_i(t).

    Candidate target sites are closest-1, closest, closest+1.
    We count candidates that are:
      - inside the microtubule,
      - untaken,
      - allowed by the neighbor-ordering constraint from the spatial code.

    This is kept only as a diagnostic. It is not used to estimate k12 in this version.
    """
    state = prc1.state

    if not prc1.is_singly_attached:
        return 0

    closest = int(prc1.closest_index_on_other_side)
    candidate_sites = [closest - 1, closest, closest + 1]

    # Same ordering bounds used in the spatial attachment routine.
    left_allowed = prc1.left_neighbor_opposite_index + 1
    right_allowed = prc1.right_neighbor_opposite_index  # exclusive

    if prc1.top_head_is_attached:
        opposite_untaken = state.bottom_untaken_sites
    elif prc1.bottom_head_is_attached:
        opposite_untaken = state.top_untaken_sites
    else:
        return 0

    count = 0
    for site in candidate_sites:
        if (
            0 <= site < state.num_sites
            and left_allowed <= site < right_allowed
            and site in opposite_untaken
        ):
            count += 1

    return count


def spatial_diagnostics(state):
    """
    Record quantities needed for the spatial -> CTMC map.

    Main mapping choices:
      - k10 is NOT inferred from p.detachment_rate. It is carried over directly later.
      - k12_i is recorded from p.double_attachment_rate for each singly bound PRC1.
      - k21_j is recorded from p.detachment_rate for each doubly bound PRC1, with no divide by 2.
    """
    singly = list(state.top_attached_prc1) + list(state.bottom_attached_prc1)
    doubly = list(state.doubly_attached_prc1)

    n1 = len(singly)
    n2 = len(doubly)
    N = n1 + n2

    # Average available sites per microtubule.
    A_spatial = 0.5 * (len(state.top_untaken_sites) + len(state.bottom_untaken_sites))

    # Diagnostic only: rough reachable available second-head target sites.
    m_values = np.array([count_second_head_targets_rough(p) for p in singly], dtype=float)

    # k10 diagnostic only: same as the spatial input per singly bound PRC1.
    # The final map sets k10_hat = singly_bound_detachment_rate directly.
    k10_values = np.full(n1, state.singly_bound_detachment_rate, dtype=float)

    # Actual spatial k12_i(t), including the distance/spring-energy correction in the code.
    # This is the total second-head attachment rate used in the spatial Gillespie rate list.
    k12_values = np.array([p.double_attachment_rate for p in singly], dtype=float)

    # Actual spatial k21_j(t), including the spring-energy correction.
    # p.detachment_rate = top_detachment_rate + bottom_detachment_rate.
    # No divide by 2 here, because the CTMC reaction n2 -> n1 uses the total event rate.
    k21_values = np.array([p.detachment_rate for p in doubly], dtype=float)

    return {
        "N": float(N),
        "n1": float(n1),
        "n2": float(n2),
        "A_spatial": float(A_spatial),
        "m_values": m_values,
        "m_sum": float(np.sum(m_values)),
        "k10_values": k10_values,
        "k10_sum": float(np.sum(k10_values)),
        "k12_values": k12_values,
        "k12_sum": float(np.sum(k12_values)),
        "k21_values": k21_values,
        "k21_sum": float(np.sum(k21_values)),
    }
```


## Spatial model runner with diagnostic recording

This is almost the same Gillespie structure as the uploaded `run_gillespie_prc1`, but the statistic function records more than just `len(state)`.


```python
def run_spatial_diagnostics_on_grid(
    initial_binding_rate_per_site,
    singly_bound_detachment_rate,
    base_double_attachment_rate,
    base_double_detachment_rate,
    times_obs,
    cooperativity_energy=0.0,
    enable_initial_attach_cooperativity=False,
    enable_second_head_attach_cooperativity=False,
    enable_second_head_detach_cooperativity=False,
    enable_hopping=False,
    max_steps=200_000,
):
    """
    Run the spatial PRC1 Gillespie model and record diagnostics on times_obs.
    """
    times_obs = np.asarray(times_obs, dtype=float)
    if times_obs.ndim != 1:
        raise ValueError("times_obs must be 1D")
    if len(times_obs) == 0:
        raise ValueError("times_obs cannot be empty")
    if np.any(np.diff(times_obs) < 0):
        raise ValueError("times_obs must be sorted")

    initial_state = make_spatial_state(
        initial_binding_rate_per_site=initial_binding_rate_per_site,
        singly_bound_detachment_rate=singly_bound_detachment_rate,
        base_double_attachment_rate=base_double_attachment_rate,
        base_double_detachment_rate=base_double_detachment_rate,
        cooperativity_energy=cooperativity_energy,
        enable_initial_attach_cooperativity=enable_initial_attach_cooperativity,
        enable_second_head_attach_cooperativity=enable_second_head_attach_cooperativity,
        enable_second_head_detach_cooperativity=enable_second_head_detach_cooperativity,
    )

    def rate_function(state):
        initial_binding_rate = state.initial_binding_rate

        if len(state) == 0:
            return [initial_binding_rate, 0, 0, 0, 0]

        rates = [initial_binding_rate] + [0] * (len(state) - 1)

        for prc1 in state:
            rates.append(prc1.double_attachment_rate)

        for prc1 in state:
            rates.append(prc1.detachment_rate)

        if enable_hopping:
            for prc1 in state:
                rates.append(prc1.total_bottom_hopping_rate)
            for prc1 in state:
                rates.append(prc1.total_top_hopping_rate)

        return rates

    def single_attach_func(state, index):
        state.single_attach_prc1()

    def double_attach_func(state, index):
        state.double_attach_prc1(index)

    def detach_func(state, index):
        state.detach_prc1(index)

    def hop_bottom_func(state, index):
        state.hop_bottom(index)

    def hop_top_func(state, index):
        state.hop_top(index)

    if enable_hopping:
        reaction_functions = [
            single_attach_func,
            double_attach_func,
            detach_func,
            hop_bottom_func,
            hop_top_func,
        ]
    else:
        reaction_functions = [
            single_attach_func,
            double_attach_func,
            detach_func,
        ]

    def statistic_function(state, stat_list, time):
        stat_list.append(spatial_diagnostics(state))

    def timestep_function(state, time, dt):
        return time + dt

    statistics, times = run_gillespie(
        initial_state,
        end_time=float(times_obs[-1]),
        rate_function=rate_function,
        reaction_functions=reaction_functions,
        statistic_function=statistic_function,
        timestep_function=timestep_function,
        max_timestep=np.inf,
        max_steps=max_steps,
    )

    # Right-continuous sampling onto times_obs.
    idx = np.searchsorted(times, times_obs, side="right") - 1
    idx = np.clip(idx, 0, len(statistics) - 1)

    scalar_keys = ["N", "n1", "n2", "A_spatial", "m_sum", "k10_sum", "k12_sum", "k21_sum"]
    out = {"times": times_obs.copy()}
    for key in scalar_keys:
        out[key] = np.array([statistics[i][key] for i in idx], dtype=float)

    # Keep the actual variable-length recorded lists too, useful for inspection.
    out["m_values_by_time"] = [statistics[i]["m_values"] for i in idx]
    out["k10_values_by_time"] = [statistics[i]["k10_values"] for i in idx]
    out["k12_values_by_time"] = [statistics[i]["k12_values"] for i in idx]
    out["k21_values_by_time"] = [statistics[i]["k21_values"] for i in idx]

    return out
```


## Saturation-aware estimator

The estimator below differs from notebook 2 in two ways:

- $\widehat{R}_{01}$ uses $n_s$ (intrinsic, empty-lattice attachment rate),
  not $\langle A^{\mathrm{spatial}}\rangle$ (which already encodes
  saturation and would double-count it in the new CTMC).
- $\widehat{k}_{12}$ divides each per-snapshot recorded
  `p.double_attachment_rate` by the local $(1 - \rho)$ at that snapshot
  before averaging, recovering the intrinsic per-molecule rate.

$\widehat{k}_{10}$ and $\widehat{k}_{21}$ are the same as in notebook 2:
the spatial input passed through, and the molecule-level
`p.detachment_rate` averaged respectively. Detachment isn't lattice-limited.


```python
def estimate_ctmc_from_spatial_runs_saturating(
    initial_binding_rate_per_site,
    singly_bound_detachment_rate,
    base_double_attachment_rate,
    base_double_detachment_rate,
    times_obs,
    n_reps=20,
    seed=1,
    verbose=True,
    n_s=None,
    **spatial_kwargs,
):
    """
    Saturation-aware mapping. Returns spatial_runs and theta_hat.

    Mapping:
      R01_hat   = 2 * k01_spatial * n_s                       # intrinsic, empty-lattice rate
      k10_hat   = singly_bound_detachment_rate                # passthrough
      k12_hat   = sum( k12_i_recorded / (1 - rho_at_t) ) / sum(n1)
      k21_hat   = sum( k21_j_recorded ) / sum(n2)             # detachment, no saturation

    n_s defaults to the number of sites per microtubule, which is the
    cap on n2 (one doubly-bound PRC1 occupies one site on each MT).
    """
    rng = np.random.default_rng(seed)
    spatial_runs = []

    for r in range(n_reps):
        np.random.seed(int(rng.integers(0, 2**31 - 1)))
        if verbose:
            print(f"Spatial run {r + 1}/{n_reps}")
        run = run_spatial_diagnostics_on_grid(
            initial_binding_rate_per_site=initial_binding_rate_per_site,
            singly_bound_detachment_rate=singly_bound_detachment_rate,
            base_double_attachment_rate=base_double_attachment_rate,
            base_double_detachment_rate=base_double_detachment_rate,
            times_obs=times_obs,
            **spatial_kwargs,
        )
        spatial_runs.append(run)

    # Determine n_s from the spatial geometry if not given.
    if n_s is None:
        # A_spatial at t=0 is the number of sites per microtubule (lattice empty).
        n_s = float(np.max([run["A_spatial"][0] for run in spatial_runs]))

    A_all       = np.concatenate([run["A_spatial"] for run in spatial_runs])
    n1_all      = np.concatenate([run["n1"]        for run in spatial_runs])
    n2_all      = np.concatenate([run["n2"]        for run in spatial_runs])
    m_sum_all   = np.concatenate([run["m_sum"]     for run in spatial_runs])
    k12_sum_all = np.concatenate([run["k12_sum"]   for run in spatial_runs])
    k21_sum_all = np.concatenate([run["k21_sum"]   for run in spatial_runs])

    total_n1 = float(np.sum(n1_all))
    total_n2 = float(np.sum(n2_all))

    # Snapshot-level occupancy fraction rho(t) = (n1 + 2*n2) / n_s
    rho_all = (n1_all + 2.0 * n2_all) / n_s
    # Numerical floor to avoid divide-by-zero if a snapshot is somehow saturated.
    one_minus_rho = np.clip(1.0 - rho_all, 1e-3, None)

    # Per-snapshot deconvolved k12 sum:
    #   k12_recorded_i = k12_intrinsic_i * (1 - rho)  =>  k12_intrinsic_i = k12_recorded_i / (1 - rho)
    # We have only the per-snapshot SUM of k12_i, so we apply the same factor
    # to the sum (valid because rho is the same for all i in that snapshot).
    k12_sum_intrinsic = k12_sum_all / one_minus_rho

    A_hat = float(np.mean(A_all))
    m_hat = float(np.sum(m_sum_all) / total_n1) if total_n1 > 0 else 0.0

    # k10, k21 unchanged from notebook 2 mapping.
    k10_hat = float(singly_bound_detachment_rate)
    k21_hat = float(np.sum(k21_sum_all) / total_n2) if total_n2 > 0 else 0.0

    # k12 with the saturation factor undone, then averaged per singly-bound molecule.
    k12_hat = float(np.sum(k12_sum_intrinsic) / total_n1) if total_n1 > 0 else 0.0

    # R01 from the empty-lattice limit (intrinsic), not from time-averaged A.
    R01_hat = float(2.0 * initial_binding_rate_per_site * n_s)

    # Diagnostics: what notebook 2's non-saturating mapping would have produced.
    R01_hat_nb2 = float(2.0 * initial_binding_rate_per_site * A_hat)
    k12_hat_nb2 = float(np.sum(k12_sum_all) / total_n1) if total_n1 > 0 else 0.0

    theta_hat = {
        "R01": R01_hat,
        "k10": k10_hat,
        "k12": k12_hat,
        "k21": k21_hat,
        "n_s": float(n_s),
        # diagnostics
        "A_hat": A_hat,
        "m_hat": m_hat,
        "R01_hat_nb2": R01_hat_nb2,
        "k12_hat_nb2": k12_hat_nb2,
        "total_n1_obs": total_n1,
        "total_n2_obs": total_n2,
    }
    return spatial_runs, theta_hat
```


## Saturating population CTMC simulator

Both attachment reactions are multiplied by $(1 - (n_1 + 2n_2)/n_s)$.
Detachment is unchanged. This is exactly the model in section 1 of the
report with all cooperativity coefficients set to zero.


```python
def simulate_population_ctmc_on_grid_saturating(theta, times_obs, seed=None):
    """
    Population CTMC with the section-1 saturation factor.
    Reactions and propensities (no cooperativity, c=0):
        a1 = R01 * (1 - (n1 + 2*n2)/n_s)               empty -> n1
        a2 = k10 * n1                                    n1 -> empty
        a3 = k12 * n1 * (1 - (n1 + 2*n2)/n_s)            n1 -> n2
        a4 = k21 * n2                                    n2 -> n1

    The (1 - .)  factor is clamped to [0, 1] for numerical safety.
    """
    R01 = float(theta["R01"])
    k10 = float(theta["k10"])
    k12 = float(theta["k12"])
    k21 = float(theta["k21"])
    n_s = float(theta["n_s"])

    rng = np.random.default_rng(seed)
    times_obs = np.asarray(times_obs, dtype=float)

    n1 = 0
    n2 = 0
    t = 0.0
    n1_out = np.zeros(len(times_obs), dtype=float)
    n2_out = np.zeros(len(times_obs), dtype=float)
    N_out  = np.zeros(len(times_obs), dtype=float)
    grid_idx = 0

    while grid_idx < len(times_obs) and times_obs[grid_idx] <= 0:
        n1_out[grid_idx] = n1
        n2_out[grid_idx] = n2
        N_out[grid_idx]  = n1 + n2
        grid_idx += 1

    while grid_idx < len(times_obs):
        sat = 1.0 - (n1 + 2.0 * n2) / n_s
        if sat < 0.0:
            sat = 0.0

        rates = np.array([
            R01 * sat,
            k10 * n1,
            k12 * n1 * sat,
            k21 * n2,
        ], dtype=float)
        a0 = rates.sum()
        if a0 <= 0:
            n1_out[grid_idx:] = n1
            n2_out[grid_idx:] = n2
            N_out[grid_idx:]  = n1 + n2
            break

        dt = rng.exponential(1.0 / a0)
        t_next = t + dt

        while grid_idx < len(times_obs) and times_obs[grid_idx] < t_next:
            n1_out[grid_idx] = n1
            n2_out[grid_idx] = n2
            N_out[grid_idx]  = n1 + n2
            grid_idx += 1

        if grid_idx >= len(times_obs):
            break

        t = t_next
        u = rng.random() * a0
        if u < rates[0]:
            n1 += 1
        elif u < rates[0] + rates[1]:
            if n1 > 0: n1 -= 1
        elif u < rates[0] + rates[1] + rates[2]:
            if n1 > 0:
                n1 -= 1
                n2 += 1
        else:
            if n2 > 0:
                n2 -= 1
                n1 += 1

    return {"times": times_obs, "n1": n1_out, "n2": n2_out, "N": N_out}


def simulate_population_ctmc_many_saturating(theta, times_obs, n_reps=20, seed=123):
    rng = np.random.default_rng(seed)
    runs = []
    for _ in range(n_reps):
        run_seed = int(rng.integers(0, 2**31 - 1))
        runs.append(simulate_population_ctmc_on_grid_saturating(theta, times_obs, seed=run_seed))
    return runs
```

## Plotting and summary helpers

This version uses two colors only, with no shaded 95\% region, and it does not export image files.


```python
def stack_runs(runs, key):
    return np.vstack([run[key] for run in runs])


def print_theta(theta_hat):
    print("Mapped CTMC parameters (saturating):")
    keys = ["R01", "k10", "k12", "k21", "n_s",
            "A_hat", "m_hat", "R01_hat_nb2", "k12_hat_nb2",
            "total_n1_obs", "total_n2_obs"]
    for key in keys:
        if key in theta_hat:
            print(f"  {key:>14s} = {theta_hat[key]:.8g}")



def _label_for_key(key):
    if key == "N":
        return r"$N(t)=n_1(t)+n_2(t)$", r"$N(t)$"
    if key == "n1":
        return r"$n_1(t)$", r"$n_1(t)$"
    if key == "n2":
        return r"$n_2(t)$", r"$n_2(t)$"
    return key, key


def plot_trajectory_and_mean(spatial_runs, ctmc_runs, times_obs, key="N", max_lines=None):
    """
    Make two plots:
      1) many trajectories from spatial and CTMC runs
      2) mean-only comparison

    Uses two colors only:
      spatial = blue
      CTMC    = orange

    No image files are exported.
    """
    spatial_mat = stack_runs(spatial_runs, key)
    ctmc_mat = stack_runs(ctmc_runs, key)

    spatial_plot = spatial_mat if max_lines is None else spatial_mat[:max_lines]
    ctmc_plot = ctmc_mat if max_lines is None else ctmc_mat[:max_lines]

    ylabel, title_name = _label_for_key(key)

    spatial_color = "tab:blue"
    ctmc_color = "tab:orange"

    # --------------------------------------------------------
    # Plot 1: many trajectories, two colors
    # --------------------------------------------------------
    fig, ax = plt.subplots(figsize=(8, 5))

    for i in range(spatial_plot.shape[0]):
        label = "Spatial trajectories" if i == 0 else None
        ax.plot(
            times_obs,
            spatial_plot[i],
            color=spatial_color,
            alpha=0.45,
            linewidth=1,
            label=label,
        )

    for i in range(ctmc_plot.shape[0]):
        label = "Mapped CTMC trajectories" if i == 0 else None
        ax.plot(
            times_obs,
            ctmc_plot[i],
            color=ctmc_color,
            alpha=0.45,
            linewidth=1,
            label=label,
        )

    ax.set_xlabel("time")
    ax.set_ylabel(ylabel)
    ax.set_title(f"Trajectory comparison for {title_name}")
    ax.legend()
    fig.tight_layout()
    plt.show()

    # --------------------------------------------------------
    # Plot 2: mean only, two colors
    # --------------------------------------------------------
    spatial_mean = spatial_mat.mean(axis=0)
    ctmc_mean = ctmc_mat.mean(axis=0)

    fig, ax = plt.subplots(figsize=(8, 5))

    ax.plot(
        times_obs,
        spatial_mean,
        color=spatial_color,
        linewidth=2.5,
        label="Spatial mean",
    )

    ax.plot(
        times_obs,
        ctmc_mean,
        color=ctmc_color,
        linewidth=2.5,
        label="Mapped CTMC mean",
    )

    ax.set_xlabel("time")
    ax.set_ylabel(ylabel)
    ax.set_title(f"Mean comparison for {title_name}")
    ax.legend()
    fig.tight_layout()
    plt.show()


def plot_n1_n2_N(spatial_runs, ctmc_runs, times_obs, max_lines=None):
    for key in ["N", "n1", "n2"]:
        plot_trajectory_and_mean(
            spatial_runs,
            ctmc_runs,
            times_obs,
            key=key,
            max_lines=max_lines,
        )
```

## User settings

Edit this cell for your test.

Default here:

- cooperativity off
- hopping off
- spring-distance energy still on
- \(k_{10}\) maps directly to `K_DET_1`
- \(k_{12}\) is estimated from actual `p.double_attachment_rate`
- \(k_{21}\) is estimated from actual `p.detachment_rate`, with no divide by 2


```python
# Observation grid (same as the report).
times_obs = np.arange(0.0, 68.0 + 4.0, 4.0)

N_SPATIAL_REPS = 20
N_CTMC_REPS = 100

# Spatial parameters (same defaults as notebook 2 so results are comparable).
INIT_BIND = 0.002       # first-head attachment per site, c_bulk folded in
K_DET_1   = 0.15        # singly-bound detachment base rate
K12_SITE  = 0.8         # second-head attachment base rate
K21_BASE  = 0.1         # double-bound detachment base rate

# Sites per microtubule. Using None -> auto-detect from A_spatial(0). Pin a
# value (e.g. 626) if you want to enforce it. With microtubule_length=5000
# and site_spacing=8, A_spatial(0) = 626.
N_S = None

SPATIAL_KWARGS = dict(
    cooperativity_energy=0.0,
    enable_initial_attach_cooperativity=False,
    enable_second_head_attach_cooperativity=False,
    enable_second_head_detach_cooperativity=False,
    enable_hopping=False,
    max_steps=200_000,
)

MAX_LINES_IN_TRAJECTORY_PLOT = None
```


## Run the trajectory check


```python
spatial_runs, theta_hat = estimate_ctmc_from_spatial_runs_saturating(
    initial_binding_rate_per_site=INIT_BIND,
    singly_bound_detachment_rate=K_DET_1,
    base_double_attachment_rate=K12_SITE,
    base_double_detachment_rate=K21_BASE,
    times_obs=times_obs,
    n_reps=N_SPATIAL_REPS,
    seed=1,
    verbose=True,
    n_s=N_S,
    **SPATIAL_KWARGS,
)

print_theta(theta_hat)

ctmc_runs = simulate_population_ctmc_many_saturating(
    theta_hat,
    times_obs,
    n_reps=N_CTMC_REPS,
    seed=2,
)

plot_n1_n2_N(
    spatial_runs,
    ctmc_runs,
    times_obs,
    max_lines=MAX_LINES_IN_TRAJECTORY_PLOT,
)
```


## Inspect recorded diagnostics

This cell checks the recorded rough quantities directly.


```python
# Mean spatial diagnostics over all spatial runs.
spatial_N = stack_runs(spatial_runs, "N")
spatial_n1 = stack_runs(spatial_runs, "n1")
spatial_n2 = stack_runs(spatial_runs, "n2")
spatial_A = stack_runs(spatial_runs, "A_spatial")
spatial_m_sum = stack_runs(spatial_runs, "m_sum")
spatial_k10_sum = stack_runs(spatial_runs, "k10_sum")
spatial_k12_sum = stack_runs(spatial_runs, "k12_sum")
spatial_k21_sum = stack_runs(spatial_runs, "k21_sum")

print("Mean N(t):", np.round(spatial_N.mean(axis=0), 3))
print("Mean n1(t):", np.round(spatial_n1.mean(axis=0), 3))
print("Mean n2(t):", np.round(spatial_n2.mean(axis=0), 3))
print("Mean A_spatial(t):", np.round(spatial_A.mean(axis=0), 3))

# These are the numerator/denominator checks used in the map.
total_m_sum = spatial_m_sum.sum()
total_n1 = spatial_n1.sum()
total_k10_sum = spatial_k10_sum.sum()
total_k12_sum = spatial_k12_sum.sum()
total_k21_sum = spatial_k21_sum.sum()
total_n2 = spatial_n2.sum()

print("\nMapping checks:")
print("sum m_i / sum n1 =", total_m_sum / total_n1 if total_n1 > 0 else np.nan)
print("old rough k12 = K12_SITE * m_hat =", K12_SITE * (total_m_sum / total_n1) if total_n1 > 0 else np.nan)
print("new actual k12 = sum k12_i / sum n1 =", total_k12_sum / total_n1 if total_n1 > 0 else np.nan)
print("k10 direct input =", K_DET_1)
print("diagnostic sum k10_i / sum n1 =", total_k10_sum / total_n1 if total_n1 > 0 else np.nan)
print("actual k21 = sum k21_j / sum n2 =", total_k21_sum / total_n2 if total_n2 > 0 else np.nan)
print("Note: k21_j here is p.detachment_rate for each doubly bound PRC1, with no divide by 2.")

# Example: show the raw variable-length arrays at one run/time.
run_id = 0
time_id = min(5, len(times_obs) - 1)
print(f"\nExample run {run_id}, time {times_obs[time_id]}:")
print("m_values:", spatial_runs[run_id]["m_values_by_time"][time_id])
print("k12_values:", spatial_runs[run_id]["k12_values_by_time"][time_id])
print("k21_values:", spatial_runs[run_id]["k21_values_by_time"][time_id])
```


## Side-by-side: saturating CTMC vs notebook-2-style non-saturating CTMC

Useful diagnostic. Same theta_hat, but we run the CTMC *without* the
$(1-\rho)$ factor and with $\widehat{R}_{01}$ recomputed from
$\langle A^{\mathrm{spatial}}\rangle$ instead of $n_s$, exactly as in
notebook 2. This isolates the effect of the saturation correction.


```python
# Build a notebook-2-style theta and simulate without saturation.
theta_nb2 = dict(theta_hat)
theta_nb2["R01"] = theta_hat["R01_hat_nb2"]
theta_nb2["k12"] = theta_hat["k12_hat_nb2"]
theta_nb2["n_s"] = float("inf")  # disables saturation in the sat-aware sim

ctmc_runs_nb2 = simulate_population_ctmc_many_saturating(
    theta_nb2, times_obs, n_reps=N_CTMC_REPS, seed=3,
)

import matplotlib.pyplot as plt
for key in ["N", "n1", "n2"]:
    spatial_mat = stack_runs(spatial_runs, key)
    sat_mat     = stack_runs(ctmc_runs, key)
    nb2_mat     = stack_runs(ctmc_runs_nb2, key)

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.plot(times_obs, spatial_mat.mean(0), color="tab:blue",   linewidth=2.5, label="Spatial mean")
    ax.plot(times_obs, sat_mat.mean(0),     color="tab:orange", linewidth=2.5, label="Saturating CTMC")
    ax.plot(times_obs, nb2_mat.mean(0),     color="tab:green",  linewidth=2.5, linestyle="--", label="Non-saturating CTMC (nb2-style)")
    ax.set_xlabel("time")
    ax.set_ylabel(key)
    ax.set_title(f"{key}: spatial vs saturating vs non-saturating")
    ax.legend()
    fig.tight_layout()
    plt.show()

print("R01 (saturating, intrinsic) =", theta_hat["R01"])
print("R01 (nb2 time-averaged)     =", theta_hat["R01_hat_nb2"])
print("k12 (saturating, intrinsic) =", theta_hat["k12"])
print("k12 (nb2 time-averaged)     =", theta_hat["k12_hat_nb2"])
print("n_s (saturating CTMC)       =", theta_hat["n_s"])
```
