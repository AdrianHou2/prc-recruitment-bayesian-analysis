# Simplified GP mapping module

Goal: train one GP inverse map

\[
(k_{12}^{\mathrm{pop}}, k_{21}^{\mathrm{pop}})
\longmapsto
(k_{12}^{\mathrm{site}}, k_{21}^{\mathrm{base}}).
\]

The expensive part is building the matched training table by running spatial simulations. After that, the GP training is short.

```python
from pathlib import Path
import importlib.util
import sys
from datetime import datetime

import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import ConstantKernel, RBF, WhiteKernel
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.multioutput import MultiOutputRegressor


def load_module(module_name, filename_candidates):
    """Load local modules, including uploaded files with parenthesized names."""
    search_roots = [
        Path.cwd(),
        Path.cwd() / "python_version",
        Path.cwd().parent,
        Path.cwd().parent / "python_version",
        Path("/mnt/data"),
    ]
    for root in search_roots:
        for name in filename_candidates:
            path = root / name
            if path.exists():
                spec = importlib.util.spec_from_file_location(module_name, path)
                mod = importlib.util.module_from_spec(spec)
                sys.modules[module_name] = mod
                spec.loader.exec_module(mod)
                print(f"Loaded {module_name} from {path}")
                return mod
    raise FileNotFoundError(f"Could not find {module_name} in {search_roots}")


# Load dependency order: prc1_state imports prc1.
gillespie = load_module("gillespie", ["gillespie.py", "gillespie(20).py", "gillespie(21).py"])
prc1 = load_module("prc1", ["prc1.py", "prc1(19).py", "prc1(20).py", "prc1(21).py"])
prc1_state = load_module("prc1_state", ["prc1_state.py", "prc1_state(18).py", "prc1_state(19).py", "prc1_state(20).py"])

from prc1_state import State
```

## Settings

```python
# Fixed spatial parameters
INIT_BIND = 0.000238   # first-head attachment rate per available site
K_DET_1   = 0.1        # singly bound detachment rate

# Observation grid
DT_OBS = 4.0
T_END = 68.0
times_obs = np.arange(0.0, T_END + DT_OBS, DT_OBS)

# Spatial geometry/constants, matching the spatial simulator defaults
MICROTUBULE_LENGTH = 5000
SITE_SPACING = 8
MICROTUBULE_OFFSET = 0
SPRING_CONSTANT = 2
REST_LENGTH = 32
KBT = 4.1
MICROTUBULE_SEPARATION = 30
BASE_HOPPING_RATE = 1
NUM_SITES = int(MICROTUBULE_LENGTH / SITE_SPACING) + 1

SPATIAL_KWARGS = dict(
    cooperativity_energy=0.0,
    enable_initial_attach_cooperativity=False,
    enable_second_head_attach_cooperativity=False,
    enable_second_head_detach_cooperativity=False,
    enable_hopping=False,
    max_steps=200_000,
)

# Training-data size. This is the expensive part.
N_TARGETS = 300
N_SPATIAL_CANDIDATES = 20
N_SPATIAL_REPS = 3

# Population target ranges
K12_POP_MIN, K12_POP_MAX = 1.0, 30.0
K21_POP_MIN, K21_POP_MAX = 0.02, 1.0

# Spatial search ranges
K12_SITE_MIN, K12_SITE_MAX = 0.1, 20.0
K21_BASE_MIN, K21_BASE_MAX = 0.005, 1.0

# Keep only reliable matched pairs for GP training.
USE_ERROR_FILTER_FOR_TRAINING = True
TRAIN_MAP_ERROR_MAX = 0.15

RANDOM_SEED = 123

OUT_DIR = Path("population_to_spatial_gp_outputs")
OUT_DIR.mkdir(exist_ok=True)
MATCH_CSV = OUT_DIR / "matched_population_to_spatial_pairs.csv"
MODEL_PATH = Path("saved_models") / "gp_module.joblib"

# If this CSV exists, the notebook can skip the expensive spatial matching step.
REUSE_MATCH_CSV_IF_AVAILABLE = True

print("times_obs:", times_obs)
print("NUM_SITES:", NUM_SITES)
print("population R01 implied at t=0:", 2.0 * INIT_BIND * NUM_SITES)
```

## Minimal helper functions

```python
def log_uniform(rng, low, high, size=None):
    return np.exp(rng.uniform(np.log(low), np.log(high), size=size))


def safe_log_ratio(a, b, eps=1e-12):
    return np.log((float(a) + eps) / (float(b) + eps))


def stack_runs(runs, key):
    return np.asarray([np.asarray(run[key], dtype=float) for run in runs], dtype=float)
```

## Spatial diagnostics and report map

```python
def state_diagnostics_for_report_map(state):
    """Diagnostics needed for the spatial -> population map at one time point."""
    doubly = list(state.doubly_attached_prc1)
    singly = list(state.top_attached_prc1) + list(state.bottom_attached_prc1)

    n1 = len(singly)
    n2 = len(doubly)
    A_spatial = 0.5 * (len(state.top_untaken_sites) + len(state.bottom_untaken_sites))

    m_sum_capped = 0.0
    k12_actual_sum = 0.0
    for p in singly:
        k12_actual_sum += float(p.double_attachment_rate)

        closest = p.closest_index_on_other_side
        if closest is None:
            continue

        count = 0
        for offset in (-1, 0, 1):
            target = int(closest + offset)
            if target < 0 or target >= state.num_sites:
                continue
            if p.top_head_is_attached and target in state.bottom_untaken_sites:
                count += 1
            elif p.bottom_head_is_attached and target in state.top_untaken_sites:
                count += 1
        m_sum_capped += min(count, 3)

    k21_sum_actual = float(sum(float(p.detachment_rate) for p in doubly))

    return dict(
        N=float(n1 + n2),
        n1=float(n1),
        n2=float(n2),
        A_spatial=float(A_spatial),
        m_sum_capped=float(m_sum_capped),
        k12_actual_sum=float(k12_actual_sum),
        k21_sum_actual=float(k21_sum_actual),
    )


def run_spatial_diagnostic_one(k12_site, k21_base, times_obs, seed=None):
    """Run one spatial trajectory and sample only the diagnostics needed for the map."""
    np.random.seed(None if seed is None else int(seed))

    state = State(
        MICROTUBULE_LENGTH,
        SITE_SPACING,
        MICROTUBULE_OFFSET,
        SPRING_CONSTANT,
        REST_LENGTH,
        KBT,
        MICROTUBULE_SEPARATION,
        INIT_BIND,
        K_DET_1,
        float(k12_site),
        float(k21_base),
        BASE_HOPPING_RATE,
        SPATIAL_KWARGS["cooperativity_energy"],
        SPATIAL_KWARGS["enable_initial_attach_cooperativity"],
        SPATIAL_KWARGS["enable_second_head_attach_cooperativity"],
        SPATIAL_KWARGS["enable_second_head_detach_cooperativity"],
    )

    enable_hopping = SPATIAL_KWARGS["enable_hopping"]
    max_steps = int(SPATIAL_KWARGS["max_steps"])
    end_time = float(np.asarray(times_obs)[-1])

    def rate_function(state):
        rates = [state.initial_binding_rate] + [0] * max(len(state) - 1, 0)
        rates += [p.double_attachment_rate for p in state]
        rates += [p.detachment_rate for p in state]
        if enable_hopping:
            rates += [p.total_bottom_hopping_rate for p in state]
            rates += [p.total_top_hopping_rate for p in state]
        return rates

    reaction_functions = [
        lambda state, index: state.single_attach_prc1(),
        lambda state, index: state.double_attach_prc1(index),
        lambda state, index: state.detach_prc1(index),
    ]
    if enable_hopping:
        reaction_functions += [
            lambda state, index: state.hop_bottom(index),
            lambda state, index: state.hop_top(index),
        ]

    event_times = [0.0]
    event_stats = [state_diagnostics_for_report_map(state)]

    t = 0.0
    for step in range(max_steps):
        if t >= end_time:
            break
        t = gillespie.gillespie_step(state, t, rate_function, reaction_functions)
        if not np.isfinite(t):
            raise RuntimeError("Non-finite time in spatial simulation")
        event_times.append(float(t))
        event_stats.append(state_diagnostics_for_report_map(state))
    else:
        raise RuntimeError("Spatial simulation hit max_steps")

    event_times = np.asarray(event_times, dtype=float)
    times_obs = np.asarray(times_obs, dtype=float)

    out = {"times": times_obs.copy()}
    for key in event_stats[0].keys():
        out[key] = np.asarray([
            event_stats[max(np.searchsorted(event_times, tt, side="right") - 1, 0)][key]
            for tt in times_obs
        ], dtype=float)
    return out


def run_spatial_diagnostic_many(k12_site, k21_base, times_obs, n_reps, seed):
    rng = np.random.default_rng(seed)
    return [
        run_spatial_diagnostic_one(
            k12_site=k12_site,
            k21_base=k21_base,
            times_obs=times_obs,
            seed=int(rng.integers(0, 2**31 - 1)),
        )
        for _ in range(int(n_reps))
    ]


def report_map_from_spatial_runs(spatial_runs, k12_site):
    """Map spatial diagnostics back to effective population parameters."""
    A_hat = float(np.mean(stack_runs(spatial_runs, "A_spatial")))
    total_n1 = float(np.sum(stack_runs(spatial_runs, "n1")))
    total_n2 = float(np.sum(stack_runs(spatial_runs, "n2")))
    total_m_sum = float(np.sum(stack_runs(spatial_runs, "m_sum_capped")))
    total_k12_actual = float(np.sum(stack_runs(spatial_runs, "k12_actual_sum")))
    total_k21_sum = float(np.sum(stack_runs(spatial_runs, "k21_sum_actual")))

    if total_n1 > 0:
        m_hat = total_m_sum / total_n1
        k12_hat_capped = float(k12_site) * m_hat
        k12_hat_actual = total_k12_actual / total_n1
    else:
        m_hat = np.nan
        k12_hat_capped = np.nan
        k12_hat_actual = np.nan

    k21_hat_actual = total_k21_sum / total_n2 if total_n2 > 0 else np.nan

    return dict(
        R01_hat=2.0 * INIT_BIND * A_hat,
        k10_hat=K_DET_1,
        k12_hat_capped=k12_hat_capped,
        k12_hat_actual_diagnostic=k12_hat_actual,
        k21_hat_actual=k21_hat_actual,
        A_hat=A_hat,
        m_hat=m_hat,
        total_n1=total_n1,
        total_n2=total_n2,
    )


def map_error_to_target(mapped, k12_pop_target, k21_pop_target):
    """Log-scale mismatch between mapped-back population pair and target population pair."""
    if not np.isfinite(mapped["k12_hat_capped"]) or not np.isfinite(mapped["k21_hat_actual"]):
        return np.inf
    e12 = safe_log_ratio(mapped["k12_hat_capped"], k12_pop_target)
    e21 = safe_log_ratio(mapped["k21_hat_actual"], k21_pop_target)
    return float(np.sqrt(e12 * e12 + e21 * e21))
```

## Build the matched dataset

```python
def generate_spatial_candidates(k12_pop_target, k21_pop_target, rng):
    """Search candidates centered around a rough inverse, plus some global proposals."""
    k12_guess = np.clip(float(k12_pop_target) / 3.0, K12_SITE_MIN, K12_SITE_MAX)
    k21_guess = np.clip(float(k21_pop_target) / 2.0, K21_BASE_MIN, K21_BASE_MAX)

    n_local = int(0.7 * N_SPATIAL_CANDIDATES)
    n_global = int(N_SPATIAL_CANDIDATES) - n_local

    candidates = []
    for _ in range(n_local):
        candidates.append((
            float(np.clip(k12_guess * np.exp(rng.normal(0.0, 0.6)), K12_SITE_MIN, K12_SITE_MAX)),
            float(np.clip(k21_guess * np.exp(rng.normal(0.0, 0.6)), K21_BASE_MIN, K21_BASE_MAX)),
        ))
    for _ in range(n_global):
        candidates.append((
            float(log_uniform(rng, K12_SITE_MIN, K12_SITE_MAX)),
            float(log_uniform(rng, K21_BASE_MIN, K21_BASE_MAX)),
        ))
    candidates.append((float(k12_guess), float(k21_guess)))
    return candidates


def evaluate_candidate(k12_pop_target, k21_pop_target, k12_site, k21_base, seed):
    try:
        spatial_runs = run_spatial_diagnostic_many(
            k12_site=k12_site,
            k21_base=k21_base,
            times_obs=times_obs,
            n_reps=N_SPATIAL_REPS,
            seed=seed,
        )
        mapped = report_map_from_spatial_runs(spatial_runs, k12_site=k12_site)
        row = dict(
            k12_site=float(k12_site),
            k21_base=float(k21_base),
            map_error=map_error_to_target(mapped, k12_pop_target, k21_pop_target),
            valid=True,
        )
        row.update(mapped)
        return row
    except Exception as e:
        return dict(
            k12_site=float(k12_site),
            k21_base=float(k21_base),
            map_error=np.inf,
            valid=False,
            error_message=str(e),
        )


def match_one_population_target(target_id, k12_pop_target, k21_pop_target):
    rng = np.random.default_rng(RANDOM_SEED + 1000 * int(target_id))
    rows = []
    for cand_id, (k12_site, k21_base) in enumerate(generate_spatial_candidates(k12_pop_target, k21_pop_target, rng)):
        row = evaluate_candidate(
            k12_pop_target=k12_pop_target,
            k21_pop_target=k21_pop_target,
            k12_site=k12_site,
            k21_base=k21_base,
            seed=RANDOM_SEED + 1000 * int(target_id) + 100 * cand_id,
        )
        row.update(dict(
            target_id=int(target_id),
            k12_pop_target=float(k12_pop_target),
            k21_pop_target=float(k21_pop_target),
        ))
        rows.append(row)

    df = pd.DataFrame(rows)
    valid = df[np.isfinite(df["map_error"])]
    return (valid if len(valid) else df).loc[(valid if len(valid) else df)["map_error"].idxmin()].to_dict()


def build_mapping_dataset():
    if REUSE_MATCH_CSV_IF_AVAILABLE and MATCH_CSV.exists():
        print("Loading existing matched dataset:", MATCH_CSV)
        return pd.read_csv(MATCH_CSV)

    rng = np.random.default_rng(RANDOM_SEED)
    targets = pd.DataFrame({
        "target_id": np.arange(N_TARGETS),
        "k12_pop_target": log_uniform(rng, K12_POP_MIN, K12_POP_MAX, size=N_TARGETS),
        "k21_pop_target": log_uniform(rng, K21_POP_MIN, K21_POP_MAX, size=N_TARGETS),
    })

    best_rows = []
    for _, target in targets.iterrows():
        target_id = int(target["target_id"])
        print(f"Target {target_id + 1}/{N_TARGETS}")
        best_rows.append(match_one_population_target(
            target_id=target_id,
            k12_pop_target=float(target["k12_pop_target"]),
            k21_pop_target=float(target["k21_pop_target"]),
        ))

    df_match = pd.DataFrame(best_rows)
    df_match.to_csv(MATCH_CSV, index=False)
    print("Saved matched dataset:", MATCH_CSV.resolve())
    return df_match


df_match = build_mapping_dataset()
print("Rows:", len(df_match))
print("Median map error:", np.nanmedian(df_match["map_error"]))
print("Good rows:", int((df_match["map_error"] <= TRAIN_MAP_ERROR_MAX).sum()), "/", len(df_match))
df_match.head()
```

## Quick check of matched data quality

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

checks = [
    ("k12_pop_target", "k12_hat_capped", "population target $k_{12}$", "mapped-back $k_{12}$"),
    ("k21_pop_target", "k21_hat_actual", "population target $k_{21}$", "mapped-back $k_{21}$"),
]

for ax, (xcol, ycol, xlabel, ylabel) in zip(axes, checks):
    good = np.isfinite(df_match[xcol]) & np.isfinite(df_match[ycol]) & np.isfinite(df_match["map_error"])
    sc = ax.scatter(df_match.loc[good, xcol], df_match.loc[good, ycol], c=df_match.loc[good, "map_error"], alpha=0.8)
    lo = min(df_match.loc[good, xcol].min(), df_match.loc[good, ycol].min())
    hi = max(df_match.loc[good, xcol].max(), df_match.loc[good, ycol].max())
    ax.plot([lo, hi], [lo, hi], "--")
    ax.set_xscale("log")
    ax.set_yscale("log")
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)

fig.colorbar(sc, ax=axes, label="map error")
fig.suptitle("Matched spatial candidates mapped back to population targets")
plt.tight_layout()
plt.show()
```

## Train the GP inverse map

```python
df_train = df_match.replace([np.inf, -np.inf], np.nan).dropna(subset=[
    "k12_pop_target", "k21_pop_target", "k12_site", "k21_base", "map_error"
]).copy()

if USE_ERROR_FILTER_FOR_TRAINING:
    df_train = df_train[df_train["map_error"] <= TRAIN_MAP_ERROR_MAX].copy()

if len(df_train) < 10:
    raise RuntimeError("Not enough good matched rows. Increase N_TARGETS/N_SPATIAL_CANDIDATES or relax TRAIN_MAP_ERROR_MAX.")

X = np.log(df_train[["k12_pop_target", "k21_pop_target"]].to_numpy(dtype=float))
Y = np.log(df_train[["k12_site", "k21_base"]].to_numpy(dtype=float))

if len(df_train) >= 20:
    idx = np.arange(len(df_train))
    train_idx, test_idx = train_test_split(idx, test_size=0.25, random_state=1)
else:
    train_idx = test_idx = np.arange(len(df_train))

X_train, X_test = X[train_idx], X[test_idx]
Y_train, Y_test = Y[train_idx], Y[test_idx]
df_test = df_train.iloc[test_idx].copy()

kernel = (
    ConstantKernel(1.0, (1e-2, 1e2))
    * RBF(length_scale=np.ones(2), length_scale_bounds=(1e-2, 1e2))
    + WhiteKernel(noise_level=1e-4, noise_level_bounds=(1e-8, 1e-1))
)

gp_model = MultiOutputRegressor(
    GaussianProcessRegressor(
        kernel=kernel,
        normalize_y=True,
        n_restarts_optimizer=1,
        random_state=2,
    )
)
gp_model.fit(X_train, Y_train)
Y_pred = gp_model.predict(X_test)

gp_r2 = r2_score(Y_test, Y_pred, multioutput="raw_values")
gp_mae = mean_absolute_error(Y_test, Y_pred, multioutput="raw_values")

print("Training rows:", len(df_train), "out of", len(df_match))
print("R2  [log k12_site, log k21_base]:", gp_r2)
print("MAE [log k12_site, log k21_base]:", gp_mae)
print("Mean MAE:", float(np.mean(gp_mae)))
print("Learned kernels:")
for name, est in zip(["k12_site", "k21_base"], gp_model.estimators_):
    print(f"  {name}:", est.kernel_)
```

## Validation plots, including the spatial-pair plot

```python
Y_true_orig = np.exp(Y_test)
Y_pred_orig = np.exp(Y_pred)
labels = ["k12_site", "k21_base"]

# Coordinate parity plots
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
for j, ax in enumerate(axes):
    true_vals = Y_true_orig[:, j]
    pred_vals = Y_pred_orig[:, j]
    lo = min(true_vals.min(), pred_vals.min())
    hi = max(true_vals.max(), pred_vals.max())
    ax.scatter(true_vals, pred_vals, alpha=0.8)
    ax.plot([lo, hi], [lo, hi], "--")
    ax.set_xscale("log")
    ax.set_yscale("log")
    ax.set_xlabel(f"matched {labels[j]}")
    ax.set_ylabel(f"GP predicted {labels[j]}")
    ax.set_title(labels[j])
plt.tight_layout()
plt.show()

# Spatial-pair plot: compare the two-dimensional pair directly.
plt.figure(figsize=(7, 6))
plt.scatter(Y_true_orig[:, 0], Y_true_orig[:, 1], alpha=0.8, label="matched spatial pair")
plt.scatter(Y_pred_orig[:, 0], Y_pred_orig[:, 1], alpha=0.8, marker="x", label="GP predicted spatial pair")

for true_pair, pred_pair in zip(Y_true_orig, Y_pred_orig):
    plt.plot([true_pair[0], pred_pair[0]], [true_pair[1], pred_pair[1]], alpha=0.2)

plt.xscale("log")
plt.yscale("log")
plt.xlabel(r"$k_{12}^{site}$")
plt.ylabel(r"$k_{21}^{base}$")
plt.title("Spatial-pair plot: matched vs GP predicted")
plt.legend()
plt.tight_layout()
plt.show()
```

## Save and use the GP module

```python
def predict_spatial_from_population(k12_pop, k21_pop, return_std=False):
    """Predict spatial parameters from population parameters on the original scale."""
    X_new = np.array([[np.log(float(k12_pop)), np.log(float(k21_pop))]], dtype=float)

    if return_std:
        means = []
        stds = []
        for est in gp_model.estimators_:
            m, s = est.predict(X_new, return_std=True)
            means.append(float(m[0]))
            stds.append(float(s[0]))
        y_log = np.asarray(means, dtype=float)
        y_log_std = np.asarray(stds, dtype=float)
        return dict(
            k12_site_pred=float(np.exp(y_log[0])),
            k21_base_pred=float(np.exp(y_log[1])),
            log_std_k12_site=float(y_log_std[0]),
            log_std_k21_base=float(y_log_std[1]),
        )

    y_log = gp_model.predict(X_new)[0]
    return dict(
        k12_site_pred=float(np.exp(y_log[0])),
        k21_base_pred=float(np.exp(y_log[1])),
    )


MODEL_PATH.parent.mkdir(exist_ok=True)

gp_module = dict(
    model=gp_model,
    model_name="gp_module",
    model_type="MultiOutputRegressor(GaussianProcessRegressor)",
    input_space="log_population",
    output_space="log_spatial",
    input_names=["k12_pop", "k21_pop"],
    output_names=["k12_site", "k21_base"],
    trained_on_n_targets=int(N_TARGETS),
    n_spatial_candidates_per_target=int(N_SPATIAL_CANDIDATES),
    n_spatial_reps=int(N_SPATIAL_REPS),
    train_map_error_max=float(TRAIN_MAP_ERROR_MAX),
    n_training_rows_after_filter=int(len(df_train)),
    k12_pop_bounds=[float(K12_POP_MIN), float(K12_POP_MAX)],
    k21_pop_bounds=[float(K21_POP_MIN), float(K21_POP_MAX)],
    k12_site_bounds=[float(K12_SITE_MIN), float(K12_SITE_MAX)],
    k21_base_bounds=[float(K21_BASE_MIN), float(K21_BASE_MAX)],
    init_bind=float(INIT_BIND),
    k_det_1=float(K_DET_1),
    spatial_kwargs=dict(SPATIAL_KWARGS),
    gp_metrics=dict(
        r2=gp_r2.tolist(),
        mae=gp_mae.tolist(),
        mean_mae=float(np.mean(gp_mae)),
    ),
    created_at=datetime.now().isoformat(timespec="seconds"),

    # Legacy aliases so older notebooks that still say CTMC can inspect metadata safely.
    legacy_input_names=["k12_ctmc", "k21_ctmc"],
    k12_ctmc_bounds=[float(K12_POP_MIN), float(K12_POP_MAX)],
    k21_ctmc_bounds=[float(K21_POP_MIN), float(K21_POP_MAX)],
)

joblib.dump(gp_module, MODEL_PATH)
print("Saved GP module to:", MODEL_PATH.resolve())
print("Example prediction:", predict_spatial_from_population(10.0, 0.2, return_std=True))
```
