# CTMC → Spatial parameter matching + ML surrogate

Goal:


a fixed population CTMC target pair

\[
(k_{12}^{\rm CTMC}, k_{21}^{\rm CTMC})
\]

is matched to spatial parameters

\[
(k_{12}^{\rm site}, k_{21}^{\rm base})
\]

by searching for spatial trajectories that reproduce the CTMC mean trajectories.
After enough matched pairs are generated, a regression model learns

\[
(\log k_{12}^{\rm CTMC}, \log k_{21}^{\rm CTMC})
\mapsto
(\log k_{12}^{\rm site}, \log k_{21}^{\rm base}).
\]

Important: bad trajectory matches are **not used as labels** for the CTMC→spatial map. They are saved as rejected/unmatched cases.

## 1. Load modules

Run this notebook from the folder containing your `python_version` directory, or put `gillespie.py`, `prc1.py`, and `prc1_state.py` in the working directory.

```python
from pathlib import Path
import importlib.util
import sys


def load_module(module_name, filename_candidates):
    search_roots = [
        Path.cwd(),
        Path.cwd() / "python_version",
        Path.cwd().parent,
        Path("/mnt/data"),
    ]
    for root in search_roots:
        for name in filename_candidates:
            path = root / name
            if path.exists():
                spec = importlib.util.spec_from_file_location(module_name, path)
                module = importlib.util.module_from_spec(spec)
                sys.modules[module_name] = module
                spec.loader.exec_module(module)
                print(f"Loaded {module_name} from {path}")
                return module
    raise FileNotFoundError(f"Could not find {module_name}; checked {search_roots}")

# Load in this order because prc1_state imports prc1.
gillespie = load_module("gillespie", ["gillespie.py", "gillespie(20).py"])
prc1 = load_module("prc1", ["prc1.py", "prc1(19).py"])
prc1_state = load_module("prc1_state", ["prc1_state.py", "prc1_state(18).py"])

from prc1_state import State
```

## 2. Fixed constants and experiment settings

`INIT_BIND` and `K_DET_1` are fixed to the paper-like values you asked for.  
The CTMC uses a fixed effective first-attachment rate

\[
R_{01}^{m CTMC}=2 k_{01}^{m site} A_{m ref}.
\]

Here `A_REF` is the reference number of available sites per microtubule. The default uses the initial number of sites from the spatial geometry.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import time

# ---------------------------------------------------------------------
# Fixed spatial / paper-like parameters
# ---------------------------------------------------------------------
INIT_BIND = 0.000238   # first-head attachment per available site
K_DET_1   = 0.1        # singly bound detachment rate

# Geometry used by the spatial model
MICROTUBULE_LENGTH = 5000
SITE_SPACING = 8
MICROTUBULE_OFFSET = 0
SPRING_CONSTANT = 2
REST_LENGTH = 32
K_B_T = 4.1
MICROTUBULE_SEPARATION = 30
BASE_HOPPING_RATE = 1

NUM_SITES = int(MICROTUBULE_LENGTH / SITE_SPACING) + 1
A_REF = float(NUM_SITES)   # initial available sites per microtubule

# Fixed CTMC parameters
R01_CTMС = 2.0 * INIT_BIND * A_REF
K10_CTMС = K_DET_1

# Observation grid
TIMES_OBS = np.arange(0.0, 68.0 + 4.0, 4.0)

print("NUM_SITES =", NUM_SITES)
print("A_REF =", A_REF)
print("R01_CTMC =", R01_CTMС)
print("K10_CTMC =", K10_CTMС)
print("times =", TIMES_OBS)
```

## 3. Computational controls

Start with the pilot settings below. After the notebook works, increase `N_TARGETS` toward 100 and/or increase the number of spatial candidates and replicates.

```python
# ---------------------------------------------------------------------
# Dataset size
# ---------------------------------------------------------------------
N_TARGETS = 100          # set to 10 first if you only want a quick test

# CTMC simulation is cheap
N_CTMC_REPS = 50

# Spatial search is expensive. Increase these for better matches.
N_SPATIAL_CANDIDATES_COARSE = 12
N_SPATIAL_REPS_COARSE = 2
N_TOP_REFINE = 3
N_SPATIAL_REPS_REFINE = 8

# Search ranges for spatial parameters
K12_SITE_MIN, K12_SITE_MAX = 0.1, 30.0
K21_BASE_MIN, K21_BASE_MAX = 0.005, 0.8

# CTMC target ranges
K12_CTMC_MIN, K12_CTMC_MAX = 1.0, 30.0
K21_CTMC_MIN, K21_CTMC_MAX = 0.02, 1.0

# Matching acceptance threshold. Smaller = stricter.
ACCEPT_ERROR = 0.15

# Trajectory error weights. Use n1 and n2 so the inverse map is less ambiguous than matching N only.
ERROR_WEIGHTS = {
    "N": 1.0,
    "n1": 1.0,
    "n2": 1.0,
}

# Spatial flags. Keep cooperativity/hopping off for the first mapping experiment.
SPATIAL_KWARGS = dict(
    cooperativity_energy=0.0,
    enable_initial_attach_cooperativity=False,
    enable_second_head_attach_cooperativity=False,
    enable_second_head_detach_cooperativity=False,
    enable_hopping=False,
    max_steps=200_000,
)

# Output files
RESULTS_CSV = Path("ctmc_to_spatial_matching_results.csv")
MODEL_FILE = Path("ctmc_to_spatial_rf_model.joblib")

GLOBAL_SEED = 1
rng_global = np.random.default_rng(GLOBAL_SEED)
```

## 4. CTMC simulator

This simulates the target population model with states \((n_1,n_2)\):

\[
\emptyset 	o n_1,\quad n_1	o \emptyset,\quad n_1	o n_2,\quad n_2	o n_1.
\]

```python
def simulate_ctmc_on_grid(R01, k10, k12, k21, times_obs, seed=None):
    rng = np.random.default_rng(seed)
    times_obs = np.asarray(times_obs, dtype=float)

    n1 = 0
    n2 = 0
    t = 0.0
    obs_idx = 0

    n1_out = np.zeros(len(times_obs), dtype=float)
    n2_out = np.zeros(len(times_obs), dtype=float)
    N_out = np.zeros(len(times_obs), dtype=float)

    def record(i):
        n1_out[i] = n1
        n2_out[i] = n2
        N_out[i] = n1 + n2

    while obs_idx < len(times_obs):
        next_obs = float(times_obs[obs_idx])

        if t >= next_obs - 1e-12:
            record(obs_idx)
            obs_idx += 1
            continue

        rates = np.array([R01, k10 * n1, k12 * n1, k21 * n2], dtype=float)
        total = rates.sum()

        if total <= 0 or not np.isfinite(total):
            t = next_obs
            continue

        dt = rng.exponential(1.0 / total)
        if t + dt >= next_obs:
            t = next_obs
            continue

        t += dt
        u = rng.random() * total

        if u < rates[0]:
            n1 += 1
        elif u < rates[0] + rates[1]:
            if n1 > 0:
                n1 -= 1
        elif u < rates[0] + rates[1] + rates[2]:
            if n1 > 0:
                n1 -= 1
                n2 += 1
        else:
            if n2 > 0:
                n2 -= 1
                n1 += 1

    return {"times": times_obs, "n1": n1_out, "n2": n2_out, "N": N_out}


def simulate_ctmc_many(R01, k10, k12, k21, times_obs, n_reps, seed):
    rng = np.random.default_rng(seed)
    runs = []
    for _ in range(int(n_reps)):
        run_seed = int(rng.integers(0, 2**31 - 1))
        runs.append(simulate_ctmc_on_grid(R01, k10, k12, k21, times_obs, seed=run_seed))
    return runs
```

## 5. Spatial simulator with diagnostics

This custom runner records `N`, `n1`, `n2`, available sites, capped second-head connectivity, and actual spatial fluxes.

```python
def make_spatial_state(k12_site, k21_base, sim_kwargs):
    return State(
        MICROTUBULE_LENGTH,
        SITE_SPACING,
        MICROTUBULE_OFFSET,
        SPRING_CONSTANT,
        REST_LENGTH,
        K_B_T,
        MICROTUBULE_SEPARATION,
        INIT_BIND,
        K_DET_1,
        float(k12_site),
        float(k21_base),
        BASE_HOPPING_RATE,
        float(sim_kwargs.get("cooperativity_energy", 0.0)),
        bool(sim_kwargs.get("enable_initial_attach_cooperativity", False)),
        bool(sim_kwargs.get("enable_second_head_attach_cooperativity", False)),
        bool(sim_kwargs.get("enable_second_head_detach_cooperativity", False)),
    )


def spatial_rate_function_factory(enable_hopping=False):
    def rate_function(state):
        initial_binding_rate = state.initial_binding_rate
        if len(state) == 0:
            if enable_hopping:
                return [initial_binding_rate, 0.0, 0.0, 0.0, 0.0]
            return [initial_binding_rate, 0.0, 0.0]

        rates = [initial_binding_rate] + [0.0] * (len(state) - 1)

        for p in state:
            rates.append(p.double_attachment_rate)
        for p in state:
            rates.append(p.detachment_rate)

        if enable_hopping:
            for p in state:
                rates.append(p.total_bottom_hopping_rate)
            for p in state:
                rates.append(p.total_top_hopping_rate)

        return rates
    return rate_function


def single_attach_func(state, index):
    state.single_attach_prc1()

def double_attach_func(state, index):
    state.double_attach_prc1(index)

def detach_func(state, index):
    state.detach_prc1(index)

def hop_bottom_func(state, index):
    state.hop_bottom(index)

def hop_top_func(state, index):
    state.hop_top(index)


def get_reaction_functions(enable_hopping=False):
    if enable_hopping:
        return [single_attach_func, double_attach_func, detach_func, hop_bottom_func, hop_top_func]
    return [single_attach_func, double_attach_func, detach_func]


def safe_spatial_step(state, time_now, rate_function, reaction_functions, max_timestep):
    num_agents = len(state) if len(state) > 0 else 1
    rates = np.asarray(rate_function(state), dtype=float)

    if rates.size == 0:
        return time_now + max_timestep
    if np.any(rates < 0) or not np.all(np.isfinite(rates)):
        raise RuntimeError(f"Invalid spatial rates: {rates}")

    total_rate = float(rates.sum())
    if total_rate <= 0:
        return time_now + max_timestep

    dt = np.random.exponential(1.0 / total_rate)
    if dt > max_timestep:
        return time_now + max_timestep

    new_time = time_now + dt
    cumulative_rates = np.cumsum(rates)
    random_value = np.random.uniform(0, 1) * total_rate
    index = int(np.searchsorted(cumulative_rates, random_value, side="right"))
    index = min(index, len(rates) - 1)

    reaction_index = index // num_agents
    agent = index % num_agents
    reaction_functions[reaction_index](state, agent)
    return new_time


def capped_reachable_sites_for_singly_bound_prc1(p):
    """
    Report-style m_i(t) cap: only offsets {-1,0,1} around the closest opposite site.
    Returns an integer in {0,1,2,3}.
    """
    state = p.state
    closest = p.closest_index_on_other_side
    if closest is None:
        return 0

    if p.bottom_head_is_attached:
        target_untaken = state.top_untaken_sites
    elif p.top_head_is_attached:
        target_untaken = state.bottom_untaken_sites
    else:
        return 0

    count = 0
    for idx in (closest - 1, closest, closest + 1):
        if 0 <= idx < state.num_sites and idx in target_untaken:
            count += 1
    return min(count, 3)


def spatial_state_snapshot(state):
    doubly = list(state.doubly_attached_prc1)
    singly = list(state.top_attached_prc1) + list(state.bottom_attached_prc1)

    n2 = len(doubly)
    n1 = len(singly)
    N = n1 + n2

    A_top = len(state.top_untaken_sites)
    A_bottom = len(state.bottom_untaken_sites)
    A_spatial = 0.5 * (A_top + A_bottom)

    m_capped_sum = float(sum(capped_reachable_sites_for_singly_bound_prc1(p) for p in singly))
    k12_flux_actual = float(sum(p.double_attachment_rate for p in singly))
    k21_flux_actual = float(sum(p.detachment_rate for p in doubly))

    return {
        "N": float(N),
        "n1": float(n1),
        "n2": float(n2),
        "A_spatial": float(A_spatial),
        "m_capped_sum": float(m_capped_sum),
        "k12_flux_actual": float(k12_flux_actual),
        "k21_flux_actual": float(k21_flux_actual),
    }


def run_spatial_diagnostic_on_grid(k12_site, k21_base, times_obs, seed, sim_kwargs=None):
    sim_kwargs = {} if sim_kwargs is None else dict(sim_kwargs)
    times_obs = np.asarray(times_obs, dtype=float)
    max_steps = int(sim_kwargs.get("max_steps", 200_000))
    enable_hopping = bool(sim_kwargs.get("enable_hopping", False))

    np.random.seed(int(seed))
    state = make_spatial_state(k12_site, k21_base, sim_kwargs)
    rate_function = spatial_rate_function_factory(enable_hopping=enable_hopping)
    reaction_functions = get_reaction_functions(enable_hopping=enable_hopping)

    out = {key: np.zeros(len(times_obs), dtype=float) for key in [
        "N", "n1", "n2", "A_spatial", "m_capped_sum", "k12_flux_actual", "k21_flux_actual"
    ]}
    out["times"] = times_obs.copy()

    t = 0.0
    obs_idx = 0
    steps = 0

    while obs_idx < len(times_obs):
        next_obs = float(times_obs[obs_idx])

        if t >= next_obs - 1e-12:
            snap = spatial_state_snapshot(state)
            for key, val in snap.items():
                out[key][obs_idx] = val
            obs_idx += 1
            continue

        if steps >= max_steps:
            raise RuntimeError("Spatial simulation hit max_steps before finishing observation grid")

        max_timestep = max(next_obs - t, 0.0)
        t = safe_spatial_step(state, t, rate_function, reaction_functions, max_timestep=max_timestep)
        steps += 1

    out["steps"] = steps
    out["k12_site"] = float(k12_site)
    out["k21_base"] = float(k21_base)
    return out


def simulate_spatial_many(k12_site, k21_base, times_obs, n_reps, seed, sim_kwargs=None):
    rng = np.random.default_rng(seed)
    runs = []
    for _ in range(int(n_reps)):
        run_seed = int(rng.integers(0, 2**31 - 1))
        runs.append(run_spatial_diagnostic_on_grid(k12_site, k21_base, times_obs, seed=run_seed, sim_kwargs=sim_kwargs))
    return runs
```

## 6. Summary, error, and plotting helpers

```python
def stack_runs(runs, key):
    return np.asarray([np.asarray(run[key], dtype=float) for run in runs], dtype=float)


def summarize_runs(runs):
    summary = {}
    for key in ["N", "n1", "n2"]:
        X = stack_runs(runs, key)
        summary[f"{key}_mean"] = X.mean(axis=0)
        summary[f"{key}_std"] = X.std(axis=0, ddof=1) if X.shape[0] > 1 else np.zeros(X.shape[1])
    for key in ["A_spatial", "m_capped_sum", "k12_flux_actual", "k21_flux_actual"]:
        if key in runs[0]:
            X = stack_runs(runs, key)
            summary[f"{key}_mean"] = X.mean(axis=0)
            summary[f"{key}_sum_total"] = float(X.sum())
    return summary


def trajectory_error(candidate_summary, target_summary, weights=None):
    weights = ERROR_WEIGHTS if weights is None else weights
    total = 0.0
    details = {}
    for key, w in weights.items():
        cand = np.asarray(candidate_summary[f"{key}_mean"], dtype=float)
        targ = np.asarray(target_summary[f"{key}_mean"], dtype=float)
        scale = max(1.0, float(np.max(np.abs(targ))))
        rmse = float(np.sqrt(np.mean(((cand - targ) / scale) ** 2)))
        details[f"rmse_{key}"] = rmse
        total += float(w) * rmse
    return float(total), details


def effective_spatial_rates_from_summary(summary, k12_site):
    n1_total = float(np.sum(summary["n1_mean"]))
    n2_total = float(np.sum(summary["n2_mean"]))
    m_total = float(np.sum(summary.get("m_capped_sum_mean", 0.0)))
    k12_flux_total = float(np.sum(summary.get("k12_flux_actual_mean", 0.0)))
    k21_flux_total = float(np.sum(summary.get("k21_flux_actual_mean", 0.0)))

    out = {}
    out["k12_capped_eff"] = float(k12_site * m_total / n1_total) if n1_total > 1e-12 else np.nan
    out["k12_actual_eff"] = float(k12_flux_total / n1_total) if n1_total > 1e-12 else np.nan
    out["k21_actual_eff"] = float(k21_flux_total / n2_total) if n2_total > 1e-12 else np.nan
    out["mean_n1_total"] = n1_total
    out["mean_n2_total"] = n2_total
    return out


def plot_mean_comparison(target_summary, spatial_summary, times_obs, title="CTMC target vs spatial match"):
    times_obs = np.asarray(times_obs, dtype=float)
    for key in ["N", "n1", "n2"]:
        plt.figure(figsize=(7, 4))
        plt.plot(times_obs, target_summary[f"{key}_mean"], linewidth=3, label="CTMC target mean")
        plt.plot(times_obs, spatial_summary[f"{key}_mean"], "--", linewidth=3, label="Spatial mean")
        plt.xlabel("time")
        plt.ylabel(key)
        plt.title(f"{title}: {key}")
        plt.legend()
        plt.tight_layout()
        plt.show()
```

## 7. Candidate generation for matching

For a CTMC target pair, the search starts near the rough inverse map

\[
k_{12}^{m site}pprox k_{12}^{m CTMC}/3,
\qquad
k_{21}^{m base}pprox k_{21}^{m CTMC}/2,
\]

then adds random local and global log-scale candidates.

```python
def clip_param_pair(k12_site, k21_base):
    k12_site = float(np.clip(k12_site, K12_SITE_MIN, K12_SITE_MAX))
    k21_base = float(np.clip(k21_base, K21_BASE_MIN, K21_BASE_MAX))
    return k12_site, k21_base


def log_uniform(rng, low, high, size=None):
    return np.exp(rng.uniform(np.log(low), np.log(high), size=size))


def generate_spatial_candidates_for_target(k12_ctmc, k21_ctmc, n_candidates, rng):
    candidates = []

    # Heuristic inverse from the report-style map.
    base_k12_site = float(k12_ctmc) / 3.0
    base_k21_base = float(k21_ctmc) / 2.0
    candidates.append(clip_param_pair(base_k12_site, base_k21_base))

    # Local log-normal perturbations around heuristic inverse.
    n_local = max(1, int(0.6 * (n_candidates - 1)))
    for _ in range(n_local):
        k12 = base_k12_site * np.exp(rng.normal(0.0, 0.7))
        k21 = base_k21_base * np.exp(rng.normal(0.0, 0.7))
        candidates.append(clip_param_pair(k12, k21))

    # Global log-uniform candidates.
    while len(candidates) < int(n_candidates):
        k12 = log_uniform(rng, K12_SITE_MIN, K12_SITE_MAX)
        k21 = log_uniform(rng, K21_BASE_MIN, K21_BASE_MAX)
        candidates.append(clip_param_pair(k12, k21))

    # De-duplicate roughly.
    unique = []
    seen = set()
    for k12, k21 in candidates:
        key = (round(k12, 6), round(k21, 6))
        if key not in seen:
            unique.append((k12, k21))
            seen.add(key)
    return unique[:int(n_candidates)]
```

## 8. Match one CTMC target to one spatial pair

This is the expensive step.

```python
def evaluate_spatial_candidate(k12_site, k21_base, target_summary, times_obs, n_reps, seed):
    try:
        runs = simulate_spatial_many(
            k12_site=k12_site,
            k21_base=k21_base,
            times_obs=times_obs,
            n_reps=n_reps,
            seed=seed,
            sim_kwargs=SPATIAL_KWARGS,
        )
        summary = summarize_runs(runs)
        err, detail = trajectory_error(summary, target_summary)
        eff = effective_spatial_rates_from_summary(summary, k12_site=k12_site)
        return {
            "valid_sim": True,
            "error": float(err),
            "summary": summary,
            "runs": runs,
            "k12_site": float(k12_site),
            "k21_base": float(k21_base),
            **detail,
            **eff,
        }
    except Exception as e:
        return {
            "valid_sim": False,
            "error": np.inf,
            "summary": None,
            "runs": None,
            "k12_site": float(k12_site),
            "k21_base": float(k21_base),
            "error_message": str(e),
        }


def match_one_ctmc_target(sample_id, k12_ctmc, k21_ctmc, seed):
    rng = np.random.default_rng(seed)

    # 1. Simulate CTMC target.
    target_runs = simulate_ctmc_many(
        R01=R01_CTMС,
        k10=K10_CTMС,
        k12=k12_ctmc,
        k21=k21_ctmc,
        times_obs=TIMES_OBS,
        n_reps=N_CTMC_REPS,
        seed=int(rng.integers(0, 2**31 - 1)),
    )
    target_summary = summarize_runs(target_runs)

    # 2. Coarse spatial search.
    candidates = generate_spatial_candidates_for_target(
        k12_ctmc=k12_ctmc,
        k21_ctmc=k21_ctmc,
        n_candidates=N_SPATIAL_CANDIDATES_COARSE,
        rng=rng,
    )

    coarse_results = []
    for j, (k12_site, k21_base) in enumerate(candidates):
        cand_seed = int(rng.integers(0, 2**31 - 1))
        result = evaluate_spatial_candidate(
            k12_site=k12_site,
            k21_base=k21_base,
            target_summary=target_summary,
            times_obs=TIMES_OBS,
            n_reps=N_SPATIAL_REPS_COARSE,
            seed=cand_seed,
        )
        result["stage"] = "coarse"
        result["candidate_id"] = j
        coarse_results.append(result)

    valid_coarse = [r for r in coarse_results if r["valid_sim"]]
    if len(valid_coarse) == 0:
        return {
            "sample_id": sample_id,
            "k12_ctmc": float(k12_ctmc),
            "k21_ctmc": float(k21_ctmc),
            "accepted": False,
            "valid_match": False,
            "best_error": np.inf,
            "message": "no valid coarse spatial simulation",
        }

    top = sorted(valid_coarse, key=lambda r: r["error"])[:N_TOP_REFINE]

    # 3. Refine top candidates with more spatial replicates.
    refine_results = []
    for j, r in enumerate(top):
        cand_seed = int(rng.integers(0, 2**31 - 1))
        result = evaluate_spatial_candidate(
            k12_site=r["k12_site"],
            k21_base=r["k21_base"],
            target_summary=target_summary,
            times_obs=TIMES_OBS,
            n_reps=N_SPATIAL_REPS_REFINE,
            seed=cand_seed,
        )
        result["stage"] = "refine"
        result["candidate_id"] = j
        refine_results.append(result)

    valid_refine = [r for r in refine_results if r["valid_sim"]]
    if len(valid_refine) == 0:
        best = min(valid_coarse, key=lambda r: r["error"])
    else:
        best = min(valid_refine, key=lambda r: r["error"])

    accepted = bool(best["error"] <= ACCEPT_ERROR)

    row = {
        "sample_id": int(sample_id),
        "k12_ctmc": float(k12_ctmc),
        "k21_ctmc": float(k21_ctmc),
        "R01_ctmc": float(R01_CTMС),
        "k10_ctmc": float(K10_CTMС),
        "best_k12_site": float(best["k12_site"]),
        "best_k21_base": float(best["k21_base"]),
        "best_error": float(best["error"]),
        "accepted": accepted,
        "valid_match": bool(best["valid_sim"]),
        "rmse_N": float(best.get("rmse_N", np.nan)),
        "rmse_n1": float(best.get("rmse_n1", np.nan)),
        "rmse_n2": float(best.get("rmse_n2", np.nan)),
        "k12_capped_eff": float(best.get("k12_capped_eff", np.nan)),
        "k12_actual_eff": float(best.get("k12_actual_eff", np.nan)),
        "k21_actual_eff": float(best.get("k21_actual_eff", np.nan)),
        "mean_n1_total": float(best.get("mean_n1_total", np.nan)),
        "mean_n2_total": float(best.get("mean_n2_total", np.nan)),
    }
    return row
```

## 9. Generate CTMC targets and build matched dataset

This cell saves after every target, so you can interrupt and keep partial results.

## 11. Train ML model: CTMC parameters → spatial parameters

Only accepted matches are used as training labels.

```python
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error
import joblib

# Use only accepted and finite rows.
df_train = df_results[df_results["accepted"].astype(bool)].copy()
for col in ["k12_ctmc", "k21_ctmc", "best_k12_site", "best_k21_base"]:
    df_train = df_train[np.isfinite(df_train[col]) & (df_train[col] > 0)]

print("training rows:", len(df_train))

if len(df_train) < 5:
    print("Not enough accepted matches to train a useful ML model. Lower ACCEPT_ERROR or generate more targets.")
else:
    X = np.column_stack([
        np.log(df_train["k12_ctmc"].to_numpy()),
        np.log(df_train["k21_ctmc"].to_numpy()),
    ])
    Y = np.column_stack([
        np.log(df_train["best_k12_site"].to_numpy()),
        np.log(df_train["best_k21_base"].to_numpy()),
    ])

    if len(df_train) >= 10:
        X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.25, random_state=1)
    else:
        X_train, X_test, Y_train, Y_test = X, X, Y, Y

    model = RandomForestRegressor(
        n_estimators=500,
        max_depth=None,
        min_samples_leaf=2,
        random_state=1,
    )
    model.fit(X_train, Y_train)

    Y_pred = model.predict(X_test)

    for j, name in enumerate(["log k12_site", "log k21_base"]):
        rmse = mean_squared_error(Y_test[:, j], Y_pred[:, j], squared=False)
        r2 = r2_score(Y_test[:, j], Y_pred[:, j]) if len(Y_test) > 1 else np.nan
        print(f"{name}: RMSE={rmse:.4f}, R2={r2:.4f}")

    joblib.dump(model, MODEL_FILE)
    print("Saved model to", MODEL_FILE)
```

## 12. Predicted vs matched spatial parameters

```python
if len(df_train) >= 5:
    X_all = np.column_stack([
        np.log(df_train["k12_ctmc"].to_numpy()),
        np.log(df_train["k21_ctmc"].to_numpy()),
    ])
    Y_all = np.column_stack([
        np.log(df_train["best_k12_site"].to_numpy()),
        np.log(df_train["best_k21_base"].to_numpy()),
    ])
    Y_hat = model.predict(X_all)

    labels = ["k12_site", "k21_base"]
    for j, label in enumerate(labels):
        true_vals = np.exp(Y_all[:, j])
        pred_vals = np.exp(Y_hat[:, j])
        lo = min(true_vals.min(), pred_vals.min())
        hi = max(true_vals.max(), pred_vals.max())

        plt.figure(figsize=(5, 5))
        plt.scatter(true_vals, pred_vals)
        plt.plot([lo, hi], [lo, hi], "--")
        plt.xscale("log")
        plt.yscale("log")
        plt.xlabel(f"matched {label}")
        plt.ylabel(f"ML predicted {label}")
        plt.title(f"Predicted vs matched {label}")
        plt.tight_layout()
        plt.show()
```

## 13. Prediction helper

Use this after training to predict spatial parameters for a new CTMC pair.

```python
def predict_spatial_from_ctmc(k12_ctmc, k21_ctmc, model=model):
    X_new = np.array([[np.log(float(k12_ctmc)), np.log(float(k21_ctmc))]], dtype=float)
    pred_log = model.predict(X_new)[0]
    k12_site_pred = float(np.exp(pred_log[0]))
    k21_base_pred = float(np.exp(pred_log[1]))
    return k12_site_pred, k21_base_pred

if len(df_train) >= 5:
    example = df_train.sort_values("best_error").iloc[0]
    k12_pred, k21_pred = predict_spatial_from_ctmc(example["k12_ctmc"], example["k21_ctmc"])
    print("Example CTMC target:")
    print("k12_ctmc =", example["k12_ctmc"])
    print("k21_ctmc =", example["k21_ctmc"])
    print("Matched spatial:")
    print("k12_site =", example["best_k12_site"])
    print("k21_base =", example["best_k21_base"])
    print("ML predicted spatial:")
    print("k12_site =", k12_pred)
    print("k21_base =", k21_pred)
```

This compares a CTMC target trajectory to a spatial trajectory using the ML-predicted spatial parameters.

```python
if len(df_train) >= 5:
    # Choose one accepted target. You can change this row.
    row = df_train.sort_values("best_error").iloc[0]

    k12_ctmc = float(row["k12_ctmc"])
    k21_ctmc = float(row["k21_ctmc"])
    k12_site_pred, k21_base_pred = predict_spatial_from_ctmc(k12_ctmc, k21_ctmc)

    target_runs = simulate_ctmc_many(
        R01=R01_CTMС,
        k10=K10_CTMС,
        k12=k12_ctmc,
        k21=k21_ctmc,
        times_obs=TIMES_OBS,
        n_reps=N_CTMC_REPS,
        seed=12345,
    )
    target_summary = summarize_runs(target_runs)

    spatial_pred_runs = simulate_spatial_many(
        k12_site=k12_site_pred,
        k21_base=k21_base_pred,
        times_obs=TIMES_OBS,
        n_reps=N_SPATIAL_REPS_REFINE,
        seed=54321,
        sim_kwargs=SPATIAL_KWARGS,
    )
    spatial_pred_summary = summarize_runs(spatial_pred_runs)

    err_pred, details_pred = trajectory_error(spatial_pred_summary, target_summary)
    print("ML predicted spatial params:")
    print("k12_site_pred =", k12_site_pred)
    print("k21_base_pred =", k21_base_pred)
    print("trajectory error =", err_pred)
    print(details_pred)

    plot_mean_comparison(target_summary, spatial_pred_summary, TIMES_OBS, title="CTMC target vs ML-predicted spatial")
```

## 15. Notes

- If very few matches are accepted, increase `N_SPATIAL_CANDIDATES_COARSE`, increase `N_TOP_REFINE`, or loosen `ACCEPT_ERROR`.
- If accepted matches are noisy, increase `N_SPATIAL_REPS_REFINE`.
- Do **not** train on rejected matches as if they were correct labels.
- Matching only `N(t)` can be non-identifiable; that is why this notebook matches `N(t)`, `n1(t)`, and `n2(t)` by default.
